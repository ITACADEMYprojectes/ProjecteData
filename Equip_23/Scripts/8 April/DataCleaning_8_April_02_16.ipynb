### Convertir columnas: texto , valores

```python
#codigos estraños
nonresponse = [-3, -4, -5, -96, -97, -98, -99]
```

```python
# convertimos variables en texto
mapping_text = {
    'b0302': 'Plan pensiones',
    'b0303': 'Fondo inver.',
    'b0304': 'Acciones',
    'b0305': 'Activo renta fija',
    'b0308': 'Cuenta o depósito ahorro',
    'b0301': 'Hipoteca',
    'b0306': 'Ptmo personal',
    'b0307': 'Tjta credito',
    'b0309': 'Seg. vida',
    'b0310': 'Seg. médico'
}

for col, text in mapping_text.items():
    df[text] = df[col].apply(
        lambda x: (
            "Sí" if x == 1 else
            "No" if x == 0 else
            x  # deixa el codi tal qual (-3, -4, -5, -96, -97, -98, -99)
        )
    )
```

```python
#convertimos codigos negativos
# diccionario de códigos negativos ECF
nonresponse_text = {
    -3:  "No sabe leer",
    -4:  "Valor no fiable",
    -5:  "Error CAPI/entrevistador",
    -96: "NS/NC",
    -97: "No sabe",
    -98: "N/A",
    -99: "No contesta"
}

# aplicamos la conversión SOLO a las columnas nuevas creadas
for text in mapping_text.values():
    df[text] = df[text].replace(nonresponse_text)
```
