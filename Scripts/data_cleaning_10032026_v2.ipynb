```python
import pandas as pd

# Diccionarios de mapeo
education_map = {1: "High School", 2: "Graduate", 3: "Postgraduate", 4: "Master/Doctor"}
day_map = {2: "Monday", 3: "Tuesday", 4: "Wednesday", 5: "Thursday", 6: "Friday"}
season_map = {1: "Summer", 2: "Autumn", 3: "Winter", 4: "Spring"}

reason_dict = {
    0: "Unknown/Other",
    1: "Infectious/Parasitic",
    2: "Neoplasms",
    3: "Blood/Immune diseases",
    4: "Endocrine/Nutritional/Metabolic",
    5: "Mental/Behavioural disorders",
    6: "Nervous system diseases",
    7: "Eye/Adnexa diseases",
    8: "Ear/Mastoid diseases",
    9: "Circulatory system diseases",
    10: "Respiratory system diseases",
    11: "Digestive system diseases",
    12: "Skin/Subcutaneous diseases",
    13: "Musculoskeletal/Connective tissue",
    14: "Genitourinary system diseases",
    15: "Pregnancy/Childbirth",
    16: "Perinatal conditions",
    17: "Congenital malformations",
    18: "Symptoms/Abnormal findings",
    19: "Injury/Poisoning",
    21: "Factors influencing health status",
    22: "Patient follow-up",
    23: "Medical consultation",
    24: "Blood donation",
    25: "Laboratory examination",
    26: "Unjustified absence",
    27: "Physiotherapy",
    28: "Dental consultation"
}

def transform_rrhh(rrhh_final):
    df = rrhh_final.copy()

    # -------------------------
    # Limpiar y convertir Work_load_Average_day a float
    # -------------------------
    if "Work_load_Average_day" in df.columns:
        df["Work_load_Average_day"] = df["Work_load_Average_day"].astype(str).str.replace(",", ".").str.strip()
        df["Work_load_Average_day"] = pd.to_numeric(df["Work_load_Average_day"], errors="coerce")
        df["Work_load_Average_day"].fillna(df["Work_load_Average_day"].mean(), inplace=True)

    # -------------------------
    # Convertir booleanos
    # -------------------------
    if "Social_drinker" in df.columns:
        df["Social_drinker"] = df["Social_drinker"].astype(bool)
    if "Social_smoker" in df.columns:
        df["Social_smoker"] = df["Social_smoker"].astype(bool)

    # -------------------------
    # Mapear Education, Day_week y Seasons si existen
    # -------------------------
    if "Education" in df.columns:
        df["Education"] = df["Education"].map(education_map).astype(str)
    if "Day_week" in df.columns:
        df["Day_week"] = df["Day_week"].map(day_map).astype(str)
    if "Seasons" in df.columns:
        df["Seasons"] = df["Seasons"].map(season_map).astype(str)

    # -------------------------
    # Crear Reason_Description basado en Reason_absence
    # -------------------------
    if "Reason_absence" in df.columns:
        df["Reason_Description"] = df["Reason_absence"].map(reason_dict)
        df["Reason_Description"].fillna("Unknown", inplace=True)
    else:
        df["Reason_Description"] = "Unknown"

    # -------------------------
    # Eliminar duplicados exactos
    # -------------------------
    df = df.drop_duplicates(keep="first")

    return df
```

```python
rrhh_final = rrhh_final.drop(columns=['Weight', 'Height'])
```

```python
rrhh_final_clean = transform_rrhh(rrhh_final)
```
