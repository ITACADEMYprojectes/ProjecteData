### Extracción de Datos Crudos

- Este script realiza la conexión con el servidor MySQL para extraer la tabla Tourist_Accommodation y almacenarla localmente en formato .csv.

- Origen: Base de Datos MySQL (Equip_22)

- Destino: Data/TouristAccommodationRaw19012026.csv

- Propósito: Servir como respaldo local "raw" antes de iniciar el proceso de limpieza.

```python
import pandas as pd
from sqlalchemy import create_engine
import os

# 1. Parámetros de conexión
host = '212.227.90.6'
database = 'Equip_22'
user = 'Equipo22'
password = 'E1q2u3i4p5o22'

# 2. Conexión con MySQL
engine = create_engine(f"mysql+mysqlconnector://{user}:{password}@{host}/{database}")
connection = engine.connect()

# 3. Cargar la tabla Tourist_Accommodation como un DataFrame
try:
    df = pd.read_sql_table('Tourist_Accommodation', connection)
    print(f"Datos cargados correctamente. Filas: {len(df)}")
except Exception as e:
    print(f"Error al cargar la tabla: {e}")
finally:
    connection.close()
    print("Conexión cerrada.")

# 4. Definición de Ruta Relativa 
nombre_archivo = "TouristAccommodationRaw19012026.csv"
ruta_relativa = os.path.join('..', 'Data', nombre_archivo)

# 5. Exportar a CSV
try:
    df.to_csv(ruta_relativa, index=False)
    print(f"Archivo guardado exitosamente en: {ruta_relativa}")
except Exception as e:
    print(f"Error al guardar el archivo: {e}")
```
