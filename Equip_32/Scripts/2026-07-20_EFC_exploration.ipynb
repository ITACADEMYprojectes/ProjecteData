```python
import pandas as pd
from pathlib import Path
```

# 1. ECF 2021


```python

data_dir = Path("../Data")
input_file = data_dir / "2026-07-20_ECF_2021_ecf_2021.dta"


df_ecf_2021 = pd.read_stata(input_file)


df_ecf_2021.head()
```

```python
lector = pd.read_stata(input_file, iterator=True)

etiquetas = lector.variable_labels()

list(etiquetas.items())[:20]

df_etiquetas = pd.DataFrame(
    etiquetas.items(),
    columns=["variable", "descripcion"]
)

display(df_etiquetas.head(30))
```

```python
del lector
```

```python
# consultar variable concreta

df_etiquetas[df_etiquetas["variable"] == "a1100"]
```

```python
# buscar por palabras

df_etiquetas[
    df_etiquetas["descripcion"]
    .str.contains(
        "hipoteca|préstamo|depósito|ahorro",
        case=False,
        na=False,
        regex=True
    )
]
```

```python
# ver solo las variables demográficas

variables_demograficas = [
    "a0000",  # género
    "a04",    # edad
    "a0100",  # país de nacimiento
    "a1030",  # estado civil
    "a1100"   # educación
]

display(df_ecf_2021[variables_demograficas].head())
```

```python
# buscar por trabajo

df_etiquetas[
    df_etiquetas["descripcion"]
    .str.contains(
        "situación laboral|trabaja|empleo|ocupación",
        case=False,
        na=False,
        regex=True
    )
]
```
