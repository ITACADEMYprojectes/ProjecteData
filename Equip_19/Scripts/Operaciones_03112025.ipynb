```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\irene\Desktop\DataAnalytics\Bootcamp\Simulacro empresarial\ProjecteData\Equip_19\Scripts\tourist_accommodation_03112025_clean.csv")
```

```python
# 1. Agregación de datos: Calcular la disponibilidad promedio por ciudad y is_instant_bookable

datos_agregados = df.groupby(['city', 'is_instant_bookable'])['availability_30'].mean().reset_index()

# 2. Transformación a formato de matriz para el heatmap
#    (Filas = City, Columnas = is_instant_bookable, Valores = Disponibilidad Promedio)
heatmap_data = datos_agregados.pivot_table(
    index='city',
    columns='is_instant_bookable',
    values='availability_30'
)

# 3. Visualización con Heatmap
plt.figure(figsize=(5, 6))
sns.heatmap(
    heatmap_data,
    annot=True,              
    fmt=".1f",               
    cmap='coolwarm',          
    linewidths=.5,           
    cbar_kws={'label': 'Tasa de Ocupación Promedio (%)'}
)
plt.title('Impacto de Reserva Inmediata en la disponibilidad media por Ciudad')
plt.ylabel('Ciudad')
plt.xlabel('Reserva Inmediata (False/True)')
plt.show()
```
