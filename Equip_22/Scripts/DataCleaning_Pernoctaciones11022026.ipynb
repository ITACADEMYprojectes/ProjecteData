```python
import pandas as pd
import numpy as np
import os
```

**Media Pernoctaciones por Comunidades Autonoma y Provincias** 

La base de datos originales se desgarcó desde la Encuesta de ocupación hotelera, del Instituto Nacional de Estadística de España (INE)

Enlace: https://www.ine.es/jaxiT3/Tabla.htm?t=2065&L=0

Titulo de la base de datos "Estancia media por comunidades autonomas y provencias"

Nombre original automáticamente de la tabla desgarcarda "2065"

```python
# Cargar datos
ruta_origen = os.path.join("..", 'Data', 'INEPernoctacionesRaw11022026.csv')

df = pd.read_csv(ruta_origen, sep=';',encoding="latin-1")
```

```python
df.info()
df.columns
```

```python
df.columns = df.columns.str.strip()
```

```python
df["Fecha"] = pd.to_datetime(
    df["Periodo"].str.replace("M", "-"),
    format="%Y-%m"
)

df["Año"] = df["Fecha"].dt.year
df["Mes"] = df["Fecha"].dt.month
```

```python
df["Mes_nombre"] = df["Fecha"].dt.month_name(locale="es_ES")
```

```python
df["media_pernoctaciones"] = (
    df["Total"]
    .astype(str)
    .str.replace(",", ".", regex=False)
)

df["media_pernoctaciones"] = pd.to_numeric(
    df["media_pernoctaciones"],
    errors="coerce"
)
```

```python
df = df.rename(columns={
    "Comunidades y Ciudades Autï¿½nomas": "Comunidades y Ciudades Autónomas"
})
```

```python
df["Comunidades y Ciudades Autónomas"] = \
    df["Comunidades y Ciudades Autónomas"].str.replace(r"^\d+\s+", "", regex=True)

df["Provincias"] = \
    df["Provincias"].str.replace(r"^\d+\s+", "", regex=True)
```

```python
df = df.drop(columns=["Periodo", "Total"])
df.info()
```

```python
# --- EXPORTACIÓN DEL DATASET LIMPIO ---

# 1. Nombre del nuevo archivo
nombre_origen_limpio = 'INEPernoctacionesClean11022026.csv'

# 2. Construimos la ruta apuntando a la misma carpeta 'Data'
# Usamos '..' para subir un nivel y luego entrar en 'Data'
ruta_origen_guardado = os.path.join('..', 'Data', nombre_origen_limpio)

# 3. Guardamos el DataFrame
# index=False evita que se cree una columna extra de números
# encoding='utf-8-sig' 
df.to_csv(ruta_origen_guardado, index=False, encoding='utf-8-sig')
```
