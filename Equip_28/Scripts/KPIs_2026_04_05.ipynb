## % empleados ausentes

```python
total_empleados = df_crudo['id'].nunique()
empleados_con_ausencia = df_crudo.loc[df_crudo['horas_ausentismo'] > 0, 'id'].nunique()
porcentaje_ausentes_crudo = (empleados_con_ausencia / total_empleados) * 100
print(f"porcentaje_ausentes: {porcentaje_ausentes_crudo:.2f}%")
```

## Mes pico de carga por ausencia

```python
suma_por_mes = df_crudo.groupby('mes_ausencia_mapeado')['horas_ausentismo'].sum()
max_val = suma_por_mes.max()
mesi_pico = suma_por_mes[suma_por_mes == max_val].index.tolist()
print("Mesi con carico massimo:", mesi_pico)
print("Ore totali nel/nei mese/i pico:", float(max_val))
```

## Motivo modal de absenteismo laboral

```python
freq = df_crudo['motivo_ausencia_mapeado'].value_counts(dropna=True)

if freq.empty:
    motivo_modal = []
    freq_motivo_modal = 0
else:
    max_freq = freq.max()
    motivo_modal = freq[freq == max_freq].index.tolist()
    freq_motivo_modal = int(max_freq)

print("Motivo modal:", motivo_modal)
print("Frequenza del motivo modal:", freq_motivo_modal)
```

## Index eficiencia relativa por carga laboral

```python
#(Suma de Hit target) / (Suma de Work load Average per day)

suma_hit_target = df_crudo['cumplimiento_objetivo'].sum()
suma_work_load = df_crudo['carga_trabajo_diaria'].sum()

if suma_work_load != 0:
    kpi = suma_hit_target / suma_work_load
    print(f"KPI: {kpi:.2f}")
else:
    print("KPI: No disponible (Work load average es zero)") 
```
