# Librerias

```python
import pandas as pd
```

# Carga de la bbdd

```python
ruta = r""
df =  pd.read_csv("ruta")
```

# KPI's:

## 1. Porcentaje de conversión a depósito

```python
# Número de clientes que suscribieron
num_suscriptores=df[df['deposit']==True].nunique().id 
print("El número de clientes que suscribieron es: ",num_suscriptores)

#Número total de clientes contactados
clientes_contactados=df.nunique().id

print("El número total de clientes contactados es: ", clientes_contactados)

#Porcentaje de conversión a depósito (kpi1)
kpi1=num_suscriptores/clientes_contactados

print("KPI 1: El porcentaje de conversión a depósito es ",kpi1*100,"%")
```

## 2. Promedio de duración de llamadas de suscriptores

```python
#Sumatoria de la duración de llamadas de suscriptores
duracion_total_llam=df[df['deposit']==True].duration.sum()
print("La duración total de las llamadas a suscriptores es:",duracion_total_llam)

#Número de clientes que suscribieron
clientes_suscritos=df[df['deposit']==True].nunique().id  

kpi2=duracion_total_llam/clientes_suscritos
print("KPI 2: La duración media de las llamadas a suscriptoreses: ",kpi2)
```

## 3. Porcentaje de llamadas a teléfono o móvil

```python
count_tlf = df[df['contact'] == 'telephone']['campaign'].sum()
count_mvl = df[df['contact'] == 'cellular']['campaign'].sum() 
count_llamadas = df['campaign'].sum()

porcentaje_tlf = (count_tlf / count_llamadas)*100
porcentaje_mvl = (count_mvl / count_llamadas)*100

kpi3_tlf = porcentaje_tlf
kpi3_mvl = porcentaje_mvl

print(f"El porcentaje de llamadas a teléfono es del {round(kpi3_tlf,2)}%.")
print(f"El porcentaje de llamadas a móvil es del {round(kpi3_mvl,2)}%.")
```

## 4. Promedio de contactos previos de suscriptores

```python
contactos_previos = df[df['deposit'] == 'yes']['previous'].sum()

promedio_contactos = (contactos_previos / num_suscriptores)

kpi4 = promedio_contactos

print(f"El promedio de contactos previos es de {round(kpi4,2)}.")
```

## 5. Mes con mayor tasa de conversión

```python
suscriptores_mes = df[df['deposit'] == 'yes'].groupby('month').count()['id']
contactados_mes = df.groupby('month').count()['id']
conversion_mes = suscriptores_mes / contactados_mes

max_mes = conversion_mes.idxmax()
max_conversion = (conversion_mes.max())*100

kpi5_mes = max_conversion
kpi5_tasa = max_conversion

print(f"El mes con mayor tasa de conversión es {kpi5_mes} con un {round(kpi5_tasa,2)}%.")
```
