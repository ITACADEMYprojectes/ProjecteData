### Valores faltantes | valores nulos por columna

```python
valoresnulos = df.isnull().sum() | df.isna().sum()
```

```python
nulosporcentaje=round(((valoresnulos / len(df)) * 100),2)

valoresfaltantes=pd.DataFrame({'nulos':valoresnulos,'Porcentaje':nulosporcentaje})


valoresfaltantes=valoresfaltantes[valoresfaltantes['nulos']> 0]

valoresfaltantes
```



Tras evaluar el dataset, se observa que los valores nulos no se distribuyen de forma aleatoria, sino que siguen patrones estructurales muy claros que responden a la propia naturaleza del negocio. Podemos clasificar los datos faltantes en cuatro grandes bloques:

1. Ausencia crítica por geografía (neighbourhood_district)

El caso más llamativo es la variable neighbourhood_district, que presenta casi un 40% de valores nulos (2.760 registros). Dado que el dataset no dispone de otras columnas geográficas de apoyo que permitan deducir o inferir el distrito faltante, nos encontramos ante una pérdida de información irrecuperable para esas filas.

Sin embargo, debido a que el 60% de los registros sí contiene esta información y resulta valiosa para el análisis espacial de los apartamentos, se descarta la eliminación de la columna. El plan de acción para la fase de limpieza será tratar esta ausencia como una categoría propia, imputando los valores nulos con la etiqueta "Desconocido" o "No especificado". Esto nos permitirá mantener intacto el volumen de datos sin falsear la información ni generar problemas técnicos en modelos posteriores.

2. El bloque de las Reseñas (Ausencia Estructural)

Las variables relacionadas con las valoraciones (first_review_date, last_review_date, reviews_per_month y las métricas review_scores_...) comparten un porcentaje de nulos casi idéntico, oscilando entre el 17.9% y el 19.1% (aprox. 1.300 registros).

Interpretación y Acción: Esto responde a una ausencia estructural lógica: representan apartamentos que aún no han recibido reservas ni valoraciones. En la limpieza, no se imputarán con medias para no falsear los datos, sino que se tratarán asignando un "0" (en cantidad de reseñas) o creando una variable indicadora ("Tiene_Reseñas").

3. Variables clave con pérdida moderada

Encontramos dos variables relevantes con un volumen intermedio de nulos:

has_availability (7.8%): Tras una inspección detallada de la estructura de esta columna, se ha detectado que no se trata de información faltante real, sino de una particularidad en el registro de los datos: el sistema original únicamente documenta explícitamente el valor "VERDADERO" (en mayúsculas) cuando el inmueble tiene disponibilidad, dejando el campo completamente en blanco (nulo) en caso contrario.
Para solucionar este problema de formato y estandarizar la información, se ha transformado la columna al tipo booleano lógico. Se ha procedido a imputar todos los valores nulos convirtiéndolos en False, mientras que los registros originales se han mapeado como True. Con esta transformación, la columna queda libre de nulos y optimizada para la fase de modelado.

price (1.8%): Al tratarse de una variable objetivo fundamental en el mercado inmobiliario, la falta de precio en 131 apartamentos requiere atención. Dependiendo del enfoque del modelo, estas filas podrían ser directamente descartadas, ya que imputar el precio podría sesgar el análisis.

4. Nulos residuales (< 0.5%)

Finalmente, un conjunto de variables descriptivas (name, description, bathrooms, bedrooms, beds, amenities_list) presenta tasas de nulos insignificantes (menos del 0.5%). Esta pérdida de datos es mínima y se considera ruido habitual en la entrada manual de datos. Su impacto es despreciable, por lo que se podrá optar por imputar la mediana/moda o eliminar directamente estas pocas filas.

## **Correlaciones**

```python
# Habia usado un pairplot pero hay dmasiadas variables. El heatmap nos permite visualizar correlaciones de forma de forma limpia y efectiva con tantas variables

plt.figure(figsize=(15, 12))
sns.heatmap(campos_numericos.corr(), cmap= 'RdBu_r', annot=True, fmt=".1f", vmin=-1, vmax=1, center=0)
plt.title("Correlaciones entre datos numéricos")
plt.show()
```

Aparecen tres bloques clave de correlaciones:

- El tamaño del alojamiento, numero de baños, etc. está directamente correlacionado con el precio. Cuanto más grande es el alojamiento (más habitaciones y camas), más sube el precio.

- La disponibilidad: availability_30, 60, 90 y 365 están casi totalmente correlacionadas entre sí. Un alojamiento que está disponible a corto plazo (30 días) es casi seguro que también estará disponible a largo plazo (365 días). Son redundantes entre ellas.

- Reviews: 
    - Las categorías de puntuación (accuracy, cleanliness, checkin, etc.) tienen valores de 0.5 a 0.8 entre sí. Efecto halo: si te ha gustado uno, tendencia a que te gusten los otros. 

    - Además, los alojamientos que han acumulado muchas reseñas a lo largo de su historia también son los que reciben reseñas más frecuentemente mes a mes. (number_of_reviews y Reviews_per_month)

    - Price y review_scores no muestra correlación, con lo cual mayor precio no garantiza mayor rating. El precio y la calidad percibida no correlacion.

- Redundancia: muchas columnas dicen lo mismo. Quizá PCA.

```python
# Ver correlación entre precio y review. El precio y la calidad percibida no correlacion.

plt.figure(figsize=(10, 6))

sns.regplot(data=df, x='review_scores_rating', y='price', 
            scatter_kws={'alpha':0.3}, 
            line_kws={'color':'red'})

plt.title('Regresión: Precio vs Número de camas')
plt.grid(True)
plt.show()
```

Correlación entre barrio y tipo de alojamiento

```python
taula_heatmap = pd.crosstab(df['neighbourhood_district'], df['room_type'], values=df['apartment_id'], aggfunc='count')
plt.figure(figsize=(15,15))
sns.heatmap(taula_heatmap, cmap='RdBu_r')
plt.show()
```

- El mercado tiene una jerarquía de precios muy clara (cuanto mayor es la capacidad, más precio)

- El rango de precios varía mucho, y un mayor precio no significa una mayor satisfacción del cliente (correlación extremadamente débil)

- El 50% de tus alojamientos tienen 2 dormitorios y 2 camas para 4 personas. Es el perfil típico de apartamento familiar o para parejas

- Las calificaciones en general son muy altas, lo que dificulta la evaluación real de alojamientos de perfil bajo, medio, alto.

- Barcelona, Madrid, Mallorca y Girona concentran la mayoría de alojamientos.

- Casi todos los alojamientos son habitaciones privadas o apartamentos enteros. No hay casi habitaciones de hoteles ni habitaciones compartidas (modelo tipo Airbnb).
