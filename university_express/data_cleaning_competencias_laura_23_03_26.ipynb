```python
EAL22 = "TABLAS_EAL_2021.xlsx"
nombre_hoja = "EAL-22"

# 1. Cargar la hoja
df_todo_22 = pd.read_excel(EAL22, sheet_name=nombre_hoja, header=None)

# --- BLOQUE 1 (Líneas 8 a 18 de Excel) ---
# Cambiado cols1 por cols1_22 para que coincida
cols1_22 = df_todo_22.iloc[7, 0:10].tolist()
df1_22 = df_todo_22.iloc[8:19, 0:10].copy()
df1_22.columns = cols1_22  # <--- AQUÍ estaba el error, ya está corregido

# --- BLOQUE 2 (Líneas 21 en adelante) ---
cols2_22 = df_todo_22.iloc[20, 0:10].tolist()
df2_22 = df_todo_22.iloc[21:32, 0:10].copy()
df2_22.columns = cols2_22 # <--- Asegúrate que aquí también sea cols2_22

# --- FUNCIÓN DE LIMPIEZA ---
def limpiar_tabla(df):
    # 1. Renombrar la primera columna
    df.rename(columns={df.columns[0]: 'Competencias'}, inplace=True)
    
    # 2. Limpiar nombres de columnas
    df.columns = [str(c).strip() for c in df.columns]
    
    # 3. Limpiar contenido de 'Competencias'
    df['Competencias'] = df['Competencias'].astype(str).str.strip()
    
    # 4. Eliminar filas no deseadas (nan y el anexo del texto largo)
    # He añadido una condición para que también ignore el texto de la nota (1)
    df = df[~df['Competencias'].str.contains('nan|Una empresa puede haber', case=False, na=False)]
    
    return df

# Aplicar limpieza
df1_22 = limpiar_tabla(df1_22)
df2_22 = limpiar_tabla(df2_22)

# --- UNIÓN FINAL ---
df_final_EAL22 = pd.merge(df1_22, df2_22, on='Competencias', how='inner')

# Mostrar resultado para verificar
print(df_final_EAL22.head())
```

```python
import pandas as pd
import os
import re

def encontrar_tabla_por_keyword(df, keyword):
    if df is None or df.empty:
        return pd.DataFrame()
    for r in range(len(df)):
        for c in range(len(df.columns)):
            valor = str(df.iloc[r, c]).upper()
            if keyword.upper() in valor:
                nueva_tabla = df.iloc[r:r+15, :].copy() 
                nueva_tabla.columns = nueva_tabla.iloc[0] 
                nueva_tabla = nueva_tabla.iloc[1:].reset_index(drop=True)
                return nueva_tabla
    return pd.DataFrame()

def limpieza_minimalista(texto):
    if pd.isna(texto): return "nan"
    t = str(texto).strip()
    if t.startswith("(1)") or "una empresa puede" in t.lower():
        return "BORRAR_NOTA"
    t = re.sub(r'\s+', ' ', t) 
    return t.strip()

todos_los_años = []

# Diccionario de mapeo (LO DEFINIMOS AQUÍ FUERA PARA QUE SIEMPRE EXISTA)
dicc_mapeo = {
    'CATALUÑA': 'CATA', 'COMUNITAT VALENCIANA': 'VALEN', 'ANDALUCÍA': 'ANDA', 
    'MADRID (COMUNIDAD DE)': 'MADRI', 'PAÍS VASCO': 'VASCO',
    'SERVICIOS': 'SERVICI', 'TOTAL': 'TOTAL', 'DE 250 A 499 TRABAJADORES': '250 A 499'
}

for anio in range(2020, 2025):
    archivo_xlsx = f"TABLAS_EAL_{anio}.xlsx"
    archivo_xls = f"TABLAS_EAL_{anio}.xls"
    archivo = archivo_xlsx if os.path.exists(archivo_xlsx) else archivo_xls if os.path.exists(archivo_xls) else None

    if not archivo: 
        print(f"⚠️ Saltando {anio}: Archivo no encontrado.")
        continue

    try:
        print(f"🔄 Procesando {archivo}...")
        df_raw_20 = pd.read_excel(archivo, sheet_name="EAL-20", header=None)
        df_raw_21 = pd.read_excel(archivo, sheet_name="EAL-21", header=None)
        df_raw_22 = pd.read_excel(archivo, sheet_name="EAL-22", header=None)

        tab_20 = encontrar_tabla_por_keyword(df_raw_20, "TOTAL")
        tab_21 = encontrar_tabla_por_keyword(df_raw_21, "SERVICIOS")
        b1 = encontrar_tabla_por_keyword(df_raw_22, "ANDALUCÍA")
        b2 = encontrar_tabla_por_keyword(df_raw_22, "VASCO")

        tablas_listas = []
        for t in [tab_20, tab_21, b1, b2]:
            if t is None or t.empty: continue
            
            t.columns = [str(c).upper().strip().replace('\n', ' ') for c in t.columns]
            t.columns = ['Competencias'] + list(t.columns[1:])
            t['Competencias'] = t['Competencias'].apply(limpieza_minimalista)
            t = t[t['Competencias'] != "BORRAR_NOTA"].copy()
            t = t.loc[:, ~t.columns.duplicated()].copy()
            tablas_listas.append(t)

        if not tablas_listas: continue

        df_temp = pd.concat(tablas_listas, axis=0, sort=False)
        df_temp['Key_Union'] = df_temp['Competencias'].str.replace(r'^\d+[\.\s\-]*', '', regex=True).str.lower().str.strip()
        
        # Agregamos numéricamente para no perder datos
        df_anio_final = df_temp.groupby('Key_Union').first().reset_index()
        df_anio_final = df_anio_final.drop(columns=['Key_Union'])

        # --- CORRECCIÓN AQUÍ: Usamos dicc_mapeo que es el que existe ---
        for estandar, clave in dicc_mapeo.items(): 
            for col in df_anio_final.columns:
                if clave in str(col).upper() and col not in ['Competencias', 'Año']:
                    df_anio_final = df_anio_final.rename(columns={col: estandar})

        df_anio_final['Año'] = anio
        
        cols_finales = ['Competencias', 'Año', 'TOTAL', 'DE 250 A 499 TRABAJADORES', 'SERVICIOS', 
                        'ANDALUCÍA', 'CATALUÑA', 'COMUNITAT VALENCIANA', 'MADRID (COMUNIDAD DE)', 'PAÍS VASCO']
        
        existentes = [c for c in cols_finales if c in df_anio_final.columns]
        todos_los_años.append(df_anio_final[existentes].copy())

    except Exception as e:
        print(f"❌ Error en {anio}: {e}")

# UNIÓN FINAL CON SEGURIDAD
if todos_los_años:
    df_final = pd.concat(todos_los_años, ignore_index=True)
    df_final = df_final[df_final['Competencias'].str.len() > 3]
    print(f"✅ ¡PROCESO COMPLETADO! Filas totales: {len(df_final)}")
    display(df_final.sort_values(['Año', 'Competencias']))
else:
    print("🚨 No se pudo procesar ninguna tabla. Verifica que los nombres de las hojas (EAL-20, etc.) sean correctos en tus archivos.")
```

```python
def limpieza_minimalista(texto):
    if pd.isna(texto): return "nan"
    t = str(texto).strip()
    if t.startswith("(1)") or "una empresa puede" in t.lower():
        return "BORRAR_NOTA"
    t = re.sub(r'\s+', ' ', t)
    return t.strip()

todos_los_años = []

dicc_mapeo = {
    'CATALUÑA': 'CATA', 'COMUNITAT VALENCIANA': 'VALEN', 'ANDALUCÍA': 'ANDA', 
    'MADRID (COMUNIDAD DE)': 'MADRI', 'PAÍS VASCO': 'VASCO',
    'SERVICIOS': 'SERVICI', 'TOTAL': 'TOTAL'
}

for anio in range(2020, 2025):
    archivo_xlsx = f"TABLAS_EAL_{anio}.xlsx"
    archivo_xls = f"TABLAS_EAL_{anio}.xls"
    archivo = archivo_xlsx if os.path.exists(archivo_xlsx) else archivo_xls if os.path.exists(archivo_xls) else None

    if not archivo: continue

    try:
        print(f"🔄 Procesando {archivo}...")
        df_raw_20 = pd.read_excel(archivo, sheet_name="EAL-20", header=None)
        df_raw_21 = pd.read_excel(archivo, sheet_name="EAL-21", header=None)
        df_raw_22 = pd.read_excel(archivo, sheet_name="EAL-22", header=None)

        tab_20 = encontrar_tabla_por_keyword(df_raw_20, "TOTAL")
        tab_21 = encontrar_tabla_por_keyword(df_raw_21, "SERVICIOS")
        b1 = encontrar_tabla_por_keyword(df_raw_22, "ANDALUCÍA")
        b2 = encontrar_tabla_por_keyword(df_raw_22, "VASCO")

        tablas_listas = []
        for i, t in enumerate([tab_20, tab_21, b1, b2]):
            if t.empty: continue
            
            # --- CORRECCIÓN CLAVE: Renombrar Índice 5 ANTES de limpiar columnas ---
            # Solo lo hacemos para la primera tabla (tab_20) que es la que tiene los tamaños
            if i == 0 and len(t.columns) >= 6:
                # Forzamos el nombre en el índice 5
                cols = list(t.columns)
                cols[5] = 'DE 250 A 499 TRABAJADORES'
                t.columns = cols

            # Limpiar nombres de columnas para que sean strings comparables
            t.columns = [str(c).upper().strip().replace('\n', ' ') for c in t.columns]
            
            # Asegurar que la primera columna siempre sea Competencias
            # (Pero mantenemos los nombres que acabamos de poner arriba)
            nombres_finales = ['Competencias'] + list(t.columns[1:])
            t.columns = nombres_finales
            
            # Limpieza de filas
            t['Competencias'] = t['Competencias'].apply(limpieza_minimalista)
            t = t[t['Competencias'] != "BORRAR_NOTA"].copy()
            
            # Crear clave de unión (normalización del texto de la competencia)
            t['Comp_Norm'] = t['Competencias'].str.replace(r'^\d+[\.\s\-]*', '', regex=True).str.strip()
            
            # Eliminar duplicados de columnas
            t = t.loc[:, ~t.columns.duplicated()].copy()
            tablas_listas.append(t)

        # Unir todas las tablas (Total, Servicios, CCAA)
        df_temp = pd.concat(tablas_listas, axis=0, sort=False)
        
        # Agrupar por el nombre normalizado para "rellenar" los huecos entre tablas
        df_anio_final = df_temp.groupby('Comp_Norm', as_index=False).first()
        df_anio_final = df_anio_final.drop(columns=['Comp_Norm'])

        # Renombrar según el diccionario de mapeo (Comunidades y Total)
        for estandar, clave in dicc_mapeo.items():
            for col in df_anio_final.columns:
                # Evitamos renombrar la de trabajadores por error
                if clave in str(col).upper() and col not in ['Competencias', 'DE 250 A 499 TRABAJADORES']:
                    df_anio_final = df_anio_final.rename(columns={col: estandar})

        df_anio_final['Año'] = anio
        
        cols_finales = ['Competencias', 'Año', 'TOTAL', 'DE 250 A 499 TRABAJADORES', 'SERVICIOS', 
                        'ANDALUCÍA', 'CATALUÑA', 'COMUNITAT VALENCIANA', 'MADRID (COMUNIDAD DE)', 'PAÍS VASCO']
        
        # Forzar que existan todas, si no existen se crean como None
        for c in cols_finales:
            if c not in df_anio_final.columns:
                df_anio_final[c] = None

        todos_los_años.append(df_anio_final[cols_finales].copy())

    except Exception as e:
        print(f"❌ Error en {anio}: {e}")

# Consolidación final de todos los años
if todos_los_años:
    df_final = pd.concat(todos_los_años, ignore_index=True)
    df_final = df_final[df_final['Competencias'].astype(str).str.len() > 3]
    display(df_final.sort_values(['Año', 'Competencias']))
```
