```python
import pandas as pd
import re

# 1. Cargar la hoja
df_18c = pd.read_excel("Tablas_EAL_2023.xlsx", sheet_name="EAL-18c", skiprows=5)
df_18c = df_18c.iloc[:, :1] # Solo nos interesa la primera columna donde está todo pegado
df_18c.columns = ["Bruto"]

# 2. Función mágica para separar el nombre del número (ej: "Andalucía75.5" -> "Andalucía", 75.5)
def separar_datos(texto):
    match = re.search(r'([a-zA-Z\s\(\),áéíóúÁÉÍÓÚñÑ]+)([\d\.]+)$', str(texto))
    if match:
        return match.group(1).strip(), float(match.group(2))
    return texto, None

# 3. Aplicar la separación
df_18c[['Categoria', 'Porcentaje_Seguridad_Salud']] = df_18c['Bruto'].apply(
    lambda x: pd.Series(separar_datos(x))
)

# 4. Limpiar y Transponer
df_18c_clean = df_18c.dropna(subset=['Porcentaje_Seguridad_Salud'])
df_18c_trans = df_18c_clean.set_index("Categoria").T

# 5. Seleccionar con los nombres exactos de tu tabla
columnas_finales = [
    "TOTAL",
    "De 250 a 499 trabajadores", 
    "Cataluña", 
    "Andalucía", 
    "Madrid (Comunidad de)", 
    "País Vasco", 
    "Comunitat Valenciana",
    "Transporte y almacenamiento"
]

# Filtramos solo las que existen para evitar errores
columnas_reales = [c for c in columnas_finales if c in df_18c_trans.columns]
eal_2023_18c = df_18c_trans[columnas_reales].copy()

# 6. Añadir año y mostrar
eal_2023_18c.insert(0, 'Año', 2023)
print("--- TABLA 18c (Limpieza por Regex) ---")
print(eal_2023_18c)
```
