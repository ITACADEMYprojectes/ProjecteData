```python
cols_to_drop = [
    # b0110
    'b0110a','b0110b','b0110c','b0110d','b0110e','b0110f','b0110g',
    'b0120a','b0120b','b0120c','b0120d','b0120e','b0120f','b0130a','b0130b','b0130c',
    # b0200 bloc
    'b0201','b0202','b0203','b0204','b0205','b0206','b0207',
    'b0208','b0209','b0210','b0212',
    'b0312','b0412','b0503','b0710a','b0710b','b0710c','b0710d','b0710e','b0710f','b0710g','b0710h','b0710i',
    # b0720
    'b0720a','b0720b','b0720c','b0720d','b0720e','b0720f','b0720g','b0720h','b0720i'
]

df = df.drop(columns=cols_to_drop, errors='ignore')
```

### Convertir columnas: texto , valores

```python
#codigos estraños
nonresponse = [-3, -4, -5, -96, -97, -98, -99]
```

```python
# convertimos variables en texto
mapping_text = {
    'b0302': 'Plan pensiones',
    'b0303': 'Fondo inver.',
    'b0304': 'Acciones',
    'b0305': 'Activo renta fija',
    'b0308': 'Cuenta o depósito ahorro',
    'b0301': 'Hipoteca',
    'b0306': 'Ptmo personal',
    'b0307': 'Tjta credito',
    'b0309': 'Seg. vida',
    'b0310': 'Seg. médico'
}

for col, text in mapping_text.items():
    df[text] = df[col].apply(
        lambda x: (
            "Sí" if x == 1 else
            "No" if x == 0 else
            x  # deixa el codi tal qual (-3, -4, -5, -96, -97, -98, -99)
        )
    )
```

```python
#convertimos codigos negativos
# diccionario de códigos negativos ECF
nonresponse_text = {
    -3:  "No sabe leer",
    -4:  "Valor no fiable",
    -5:  "Error CAPI/entrevistador",
    -96: "NS/NC",
    -97: "No sabe",
    -98: "N/A",
    -99: "No contesta"
}

# aplicamos la conversión SOLO a las columnas nuevas creadas
for text in mapping_text.values():
    df[text] = df[text].replace(nonresponse_text)
```

```python
df=df.drop(columns=['b0301','b0302','b0303','b0304','b0305','b0306','b0307','b0308','b0309','b0310'])
```
