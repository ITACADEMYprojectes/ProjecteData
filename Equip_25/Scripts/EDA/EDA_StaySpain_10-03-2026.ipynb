### Descripción del dataset

El dataset tiene un total de 7001 registros y 36 columnas:

- **apartment_id**: id del alojamiento turístico. 
- **name**: nombre del alojamiento.
- **description**: Breve descripción del alojamiento.
- **host_id**: id del cliente.
- **neighbourhood_name**: Nombre del pueblo o barrio, si es una ciudad, donde se encuneta el alojamiento.
- **neighbourhood_district**: Nombre del distrito del barrio, donde está ubicado el alojamiento.
- **room_type**: Modalidad de alojamiento (apartamento completo o habitación provada).
- **accommodates**: Plazas que alberga el alojamiento.
- **bathrooms**: Número de baños que tiene el alojamiento.
- **bedrooms**: Número de habitaciones que tiene el alojamiento.
- **beds**: Número de camas que tiene el alojamiento.
- **amenities_list**: Lista de los servicios y equipamineto del que dispone el alojamiento.
- **price**: Precio del alquiler del alojamiento.
- **minimum_nights**: Número mínimo de noches que se puede alquilar el alojaimento.
- **maximum_nights**: Número máximo de noches que se puede alquilar el alojaimento.
- **has_availability**: Indica si el alojamiento tiene fechas disponibles para ser reservado (Verdadero/Falso).
- **availability_30**: Número noches disponibles en los próximos 30 días.
- **availability_60**: Número noches disponibles en los próximos 60 días.
- **availability_90**: Número noches disponibles en los próximos 90 días.
- **availability_365**: Número noches disponibles en los próximos 365 días.
- **number_of_reviews**: Número de reseñas recibidas.
- **first_review_date**: Fecha de la primera reseña.
- **last_review_date**: Fecha de la última reseña.
- **review_scores_rating**: Puntuación media de los huéspedes sobre el alojamiento.
- **review_scores_accuracy**: Puntuación media de los huespedes sobre la veracidad del anuncio a la realidad.
- **review_scores_cleanliness**: Puntuación media de los huespedes sobre la limpieza del alojamiento.
- **review_scores_checkin**: Puntuación media de los huespedes sobre la entrada y acceso  al alojamineto.
- **review_scores_communication**: Puntuación media de los huespedes sobre la comunicación con el propietario del alojamiento.
- **review_scores_location**: Puntuación media de los huespedes sobre la ubicación del alojamiento.
- **review_scores_value**:Puntuación media de los huespedes sobre la relación calidad - precio.
- **is_instant_bookable**: Indica si el alojamiento permite la reserva automática sin confirmación previa del anfitrión (VERDADERO/FALSO).
- **reviews_per_month**: Número de reseñas recibidas cada mes.
- **country**: País donde se encuentra el alojamineto.
- **city**: Ciudad, o província, donde se encunetra el alojamiento.
- **insert_date**: Fecha desde la que está publicado el alojamiento.

## **Correlaciones**

```python
# Habia usado un pairplot pero hay dmasiadas variables. El heatmap nos permite visualizar correlaciones de forma de forma limpia y efectiva con tantas variables

plt.figure(figsize=(15, 12))
sns.heatmap(campos_numericos.corr(), cmap= 'RdBu_r', annot=True, fmt=".1f", vmin=-1, vmax=1, center=0)
plt.title("Correlaciones entre datos numéricos")
plt.show()
```

Aparecen tres bloques clave de correlaciones:

- El tamaño del alojamiento, numero de baños, etc. está directamente correlacionado con el precio. Cuanto más grande es el alojamiento (más habitaciones y camas), más sube el precio.

- La disponibilidad: availability_30, 60, 90 y 365 están casi totalmente correlacionadas entre sí. Un alojamiento que está disponible a corto plazo (30 días) es casi seguro que también estará disponible a largo plazo (365 días). Son redundantes entre ellas.

- Reviews: 
    - Las categorías de puntuación (accuracy, cleanliness, checkin, etc.) tienen valores de 0.5 a 0.8 entre sí. Efecto halo: si te ha gustado uno, tendencia a que te gusten los otros. 

    - Además, los alojamientos que han acumulado muchas reseñas a lo largo de su historia también son los que reciben reseñas más frecuentemente mes a mes. (number_of_reviews y Reviews_per_month)

    - Price y review_scores no muestra correlación, con lo cual mayor precio no garantiza mayor rating. El precio y la calidad percibida no correlacion.

- Redundancia: muchas columnas dicen lo mismo. Quizá PCA.

```python
# Ver correlación entre precio y review. El precio y la calidad percibida no correlacion.

plt.figure(figsize=(10, 6))

sns.regplot(data=df, x='review_scores_rating', y='price', 
            scatter_kws={'alpha':0.3}, 
            line_kws={'color':'red'})

plt.title('Regresión: Precio vs Número de camas')
plt.grid(True)
plt.show()
```

Correlación entre barrio y tipo de alojamiento

```python
taula_heatmap = pd.crosstab(df['neighbourhood_district'], df['room_type'], values=df['apartment_id'], aggfunc='count')
plt.figure(figsize=(15,15))
sns.heatmap(taula_heatmap, cmap='RdBu_r')
plt.show()
```

## **Conclusiones y resumen del EDA**

- El mercado tiene una jerarquía de precios muy clara (cuanto mayor es la capacidad, más precio)

- El rango de precios varía mucho, y un mayor precio no significa una mayor satisfacción del cliente (correlación extremadamente débil)

- El 50% de tus alojamientos tienen 2 dormitorios y 2 camas para 4 personas. Es el perfil típico de apartamento familiar o para parejas

- Las calificaciones en general son muy altas, lo que dificulta la evaluación real de alojamientos de perfil bajo, medio, alto.

- Barcelona, Madrid, Mallorca y Girona concentran la mayoría de alojamientos.

- Casi todos los alojamientos son habitaciones privadas o apartamentos enteros. No hay casi habitaciones de hoteles ni habitaciones compartidas (modelo tipo Airbnb).
