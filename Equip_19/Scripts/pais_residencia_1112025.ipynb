```python
# Solo españoles y extranjeros
df_nac_internac = df.loc[(df['pais_residencia'] == 'España') | df['pais_residencia'] == 'No residentes en España']

# Solo países o regiones
df_paises = df.loc[(df['pais_residencia'] != 'Total') & (df['pais_residencia'] != 'No residentes en España') 
                   & (df['pais_residencia'] != 'Unión Europea (sin España)') & (df['pais_residencia'] != 'Extranjeros')]
```

```python
df_paises_agrupado = df_paises.groupby('pais_residencia').agg(
    Suma_Viajeros=('total_viajeros', 'sum'),
    Media_Pernoctaciones=('pernoctaciones', 'mean') 
)
```
