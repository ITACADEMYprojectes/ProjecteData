### EVALUAR RESPUESTAS x cada colummna

```python
### FUNCIÓN evaluar_ TEST DE LA FUNCIÓN
def evaluar_correcta_simple(serie, valores_correctos):
    """Devuelve 1 si el valor está en valores_correctos, 0 si no, NaN si la celda es NaN."""
    return serie.where(serie.isna(), serie.isin(valores_correctos).astype(int))
```

```python
# Prueba: serie con un valor correcto (3), incorrectos (1, 2) y un NaN
serie_prueba = pd.Series([3, 1, np.nan, 4, 2])
display(evaluar_correcta_simple(serie_prueba, [3, 4])) ## valores correctos 3, 4
```

```python
### FUNCIÓN  

def evaluar_interes_compuesto(serie_compuesto, serie_correcta_simple):
    """Correcto (1) solo si el valor es 1 y además interes-simple fue correcta. NaN si el valor es NaN."""
    return serie_compuesto.where(
        serie_compuesto.isna(),
        ((serie_compuesto == 1) & (serie_correcta_simple == 1)).astype(int)
    )
```

```python
# Prueba: compuesto correcto+simple correcta, compuesto correcto+simple incorrecta (dependencia),
# compuesto incorrecto, y compuesto NaN
serie_compuesto = pd.Series([1, 1, 2, np.nan])
serie_simple_correcta = pd.Series([1, 0, 1, 1])
display(evaluar_interes_compuesto(serie_compuesto, serie_simple_correcta))
```

```python
### EVALUACIÓN POR COLUMNA

df["correcta_division_simple"] = evaluar_correcta_simple(df["division_simple"], [200])
df["correcta_inflacion"] = evaluar_correcta_simple(df["inflacion"], [3, 4])
df["correcta_interes_prestamo"] = evaluar_correcta_simple(df["interes-prestamo"], [0])
df["correcta_interes_simple"] = evaluar_correcta_simple(df["interes-simple"], [102])
df["correcta_riesgo_a"] = evaluar_correcta_simple(df["riesgo_a"], [1])
df["correcta_riesgo_b"] = evaluar_correcta_simple(df["riesgo_b"], [1])
df["correcta_riesgo_c"] = evaluar_correcta_simple(df["riesgo_c"], [1])
df["correcta_riesgo_d"] = evaluar_correcta_simple(df["riesgo_d"], [1])
```

```python
#### aplicar interes-compuesto, que depende de correcta_interes_simple ya calculada en la celda anterior

df["correcta_interes_compuesto"] = evaluar_interes_compuesto(df["interes-compuesto"], df["correcta_interes_simple"])
```

```python
## función CALCULAR SCORE

def calcular_score(df, columnas):
    """Suma las columnas indicadas fila por fila, tratando NaN como 0."""
    return df[columnas].fillna(0).sum(axis=1).astype(int)
```

```python
## función CLASIFICAR
def clasificar_ranking(score):
    """Clasifica el score 0-7 en bajo (0-3), medio (4-5) o alto (6-7)."""
    if score <= 3:
        return "bajo"
    elif score <= 5:
        return "medio"
    return "alto"
```

### Rangos por edad

18-30 : jovenes, 
31-40 : adultos-jovenes, 
41-50 : adultos, 
51-60 : senior, 
61-80 : senior-mayores

```python
## función pandas cut, cortar por rangos

def asignar_age_bracket(serie_edad):
    """Clasifica la edad en bracket: jovenes, adultos-jovenes, adultos, senior, mayores. NaN si la edad es NaN."""
    bordes = [17, 30, 40, 50, 60, 80]
    etiquetas = ["jovenes", "adultos-jovenes", "adultos", "senior", "mayores"]
    return pd.cut(pd.to_numeric(serie_edad, errors="coerce"), bins=bordes, labels=etiquetas)


## columna rango-edad
df["rango_edad"] = asignar_age_bracket(df["edad"])

## ver
display(df[["edad", "rango_edad"]].head(10))
```

### Porcentajes por bracket de edad: 
dentro de cada grupo de edad, qué proporción es alto/medio/bajo (cada barra suma 100%).

```python
## función group by personas por cada combinación de rango_edad y ranking_conocimiento
def contar_por_bracket_y_ranking(df):
    """Cuenta personas por cada combinación de rango_edad y ranking_conocimiento."""
    return df.groupby(["rango_edad", "ranking_conocimiento"]).size().reset_index(name="cantidad")



### función convertir la cantidad en porcentaje  
def calcular_porcentaje_por_bracket(conteo):
    """Convierte la cantidad en porcentaje dentro de cada rango_edad (cada bracket suma 100%)."""
    return conteo.assign(porcentaje=conteo["cantidad"] / conteo.groupby("rango_edad")["cantidad"].transform("sum") * 100)


## aplicar funciones
bracket_ranking_porcentaje = calcular_porcentaje_por_bracket(contar_por_bracket_y_ranking(df))


## gráfico BARRAS
fig_rangos_edad = px.bar(
    bracket_ranking_porcentaje,
    x="rango_edad",
    y="porcentaje",
    color="ranking_conocimiento",
    color_discrete_map=colores_ranking,
    category_orders={"ranking_conocimiento": ["bajo", "medio", "alto"]},
    text="porcentaje",
    )
fig_rangos_edad.update_traces(
    textposition="auto",
    textfont=dict(size=20, color="#050a30", ),
    )
fig_rangos_edad.update_traces(texttemplate="%{text:.0f}%")
fig_rangos_edad.show()
```

```python
def resumen_score_por_bracket(df):
    """Calcula media, mediana y desviación estándar del score por cada rango de edad."""
    return df.groupby("rango_edad")["score_conocimiento_0_7"].agg(["mean", "median", "std", "count"])

display(resumen_score_por_bracket(df))
```

```python
## función

def media_score_por_bracket(df):
    """Calcula la media del score de conocimiento financiero por cada rango de edad."""
    return df.groupby("rango_edad")["score_conocimiento_0_7"].mean().reset_index(name="score_promedio")

## orden grupo de edades
orden_brackets = ["jovenes", "adultos-jovenes", "adultos", "senior", "mayores"]

media_bracket = media_score_por_bracket(df).assign(
    rango_edad=lambda d: pd.Categorical(d["rango_edad"], categories=orden_brackets, ordered=True)
).sort_values("rango_edad")


#gráfico
fig_line = px.line(
    media_bracket,
    x="rango_edad",
    y="score_promedio",
    markers=True,
    title="Score promedio de conocimiento financiero por rango de edad",
    text="score_promedio",
    
)
fig_line.update_yaxes(range=[3.5, 4.5])
fig_line.update_traces(texttemplate="%{text:.2f}", textfont_size=16,  )
fig_line.show()
```

#### Analizar en detalle son: inflacion, interes-compuesto, riesgo_d, 
columna de acierto: correcta_inflacion, correcta_interes_compuesto, correcta_riesgo_d.

```python
### distribución de acierto por rango_edad


## función para calcular tasa de acierto por bracket
def tasa_acierto_por_bracket(df, columna_correcta):
    """Calcula el porcentaje de acierto de una columna correcta_* dentro de cada rango_edad."""
    return df.groupby("rango_edad")[columna_correcta].mean().mul(100).reset_index(name="porcentaje_acierto")


## unir en un  dataframe
variables_detalle = ["correcta_inflacion", "correcta_interes_compuesto", "correcta_riesgo_d"]


acierto_por_bracket = pd.concat(
    [tasa_acierto_por_bracket(df, col).assign(variable=col) for col in variables_detalle],
    ignore_index=True,
)

display(acierto_por_bracket.head(20))
```

```python
## gráfico de conceptos financieros

orden_brackets = ["jovenes", "adultos-jovenes", "adultos", "senior", "mayores"]

nombres_variables = {
    "correcta_inflacion": "Inflación",
    "correcta_interes_compuesto": "Interés compuesto",
    "correcta_riesgo_d": "Diversificación de riesgo",
}

colores_variables = {
    "Inflación": "powderblue",
    "Interés compuesto": "thistle",
    "Diversificación de riesgo": "khaki",
}

acierto_por_bracket_ordenado = acierto_por_bracket.assign(
    rango_edad=lambda d: pd.Categorical(d["rango_edad"], categories=orden_brackets, ordered=True),
    variable=lambda d: d["variable"].map(nombres_variables),
).sort_values("rango_edad")


##grafico
fig_detalle = px.bar(
    acierto_por_bracket_ordenado,
    x="rango_edad",
    y="porcentaje_acierto",
    color="variable",
    color_discrete_map=colores_variables,
    category_orders={"variable": ["Inflación", "Diversificación de riesgo", "Interés compuesto"]},
    barmode="group",
    title="Porcentaje de acierto por conceptos financieros",
    labels={
        "rango_edad": "Rango de edad",
        "porcentaje_acierto": "Porcentaje de acierto",
        "variable": "Concepto financiero",
        },
    text="porcentaje_acierto",
)
fig_detalle.update_traces(texttemplate="%{text:.0f}%", textfont_size=16,)
fig_detalle.show()
```
