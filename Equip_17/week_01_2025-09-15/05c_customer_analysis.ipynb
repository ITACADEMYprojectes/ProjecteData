# Quin percentatge d'allotjaments tenen una valoració general més gran de 80 a cada ciutat?

```python
import pandas as pd

# Creamos una columna nueva que será TRUE si el review_scores_rating es superior a 80
df["rating_greather_than_80"] = df["review_scores_rating"] > 80

df.head()
```

```python
# Agrupamos por ciudad y calculamos el porcentaje
df_grouped = (df.groupby("city")["rating_greather_than_80"].mean() * 100).round(2).reset_index().sort_values(by="rating_greather_than_80", ascending=False)

df_grouped
```
