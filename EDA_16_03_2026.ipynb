```python
RRHH['Work_load_Average_day'] = RRHH['Work_load_Average_day'].astype(str).str.replace(',', '.').astype(float) # he cambiado para tipo float para sacar el heatmap

cols_to_fix = ['Disciplinary_failure', 'Education', 'Social_drinker', 'Social_smoker', 'Son', 'Pet']
for col in cols_to_fix:
    RRHH[col] = pd.to_numeric(RRHH[col], errors='coerce')


corr = RRHH.select_dtypes(include=[np.number]).corr()


plt.figure(figsize=(16, 10))
sns.heatmap(corr, 
            annot=True,      
            fmt=".2f",      
            cmap='RdYlGn',   
            vmin=-1,        
            vmax=1,          
            center=0,      
            linewidths=.5)   

plt.title('Matriz de Correlación Operativa - Dataset Equip_24', fontsize=16, loc='left')
plt.show()
```

Existe una correlación superior al 90% entre el Peso (Weight) y el Índice de Masa Corporal (BMI). 
Dado que el BMI es un indicador calculado directamente a partir del peso y la altura, estas variables 
presentan una alta redundancia (multicolinealidad). Por lo tanto, resulta estratégicamente viable 
eliminar las columnas de peso y altura para trabajar exclusivamente con el BMI, simplificando el 
modelo sin perder información crítica sobre el estado de salud de los empleados.

Se detecta una correlación moderada entre la antigüedad (Service_time) y la edad (Age) de 0.66, 
lo que es esperable dado que los empleados con mayor experiencia en la empresa tienden a ser de 
mayor edad. Adicionalmente, ambas variables presentan una correlación positiva con el BMI 
(0.47 y 0.42 respectivamente), lo que sugiere que el segmento de mayor experiencia puede presentar 
mayores riesgos de salud asociados.

Se observa una correlación negativa de -0.51 entre Reason_absence y Disciplinary_failure, lo que 
indica que los empleados con mayor número de faltas disciplinarias tienden a ausentarse por razones 
distintas a causas médicas justificadas.

La variable Hit_target presenta una correlación negativa de -0.44 con Month_absence, lo que sugiere 
que en los meses con mayor absentismo el cumplimiento de objetivos tiende a ser menor, evidenciando 
el impacto directo de las ausencias sobre el rendimiento operativo del equipo.

Se detecta una nueva correlación relevante respecto al dataset anterior: Month_absence presenta una 
correlación positiva de 0.43 con Seasons, lo que es coherente con la relación natural entre el mes 
del año y la estación, y podría implicar redundancia entre ambas variables a evaluar en la fase de 
limpieza.

```python
plt.figure(figsize=(14, 6))

mean_hours = RRHH.groupby('Reason_absence')['Absenteeism_hours'].mean().sort_values(ascending=False)

sns.barplot(x=mean_hours.index, y=mean_hours.values, palette='rocket')
plt.title('Media de Horas de Ausencia por Motivo', loc='left')
plt.xlabel('Reason_absence')
plt.ylabel('Media de horas')
plt.xticks(rotation=45)
sns.despine()
plt.tight_layout()
plt.show()
```

El barplot de la media de horas de ausencia por motivo revela una diferencia muy pronunciada 
entre las ausencias por enfermedad grave y las ausencias administrativas o de seguimiento médico.

Los motivos con mayor impacto en horas son los vinculados a enfermedades CID: el código 9 
(enfermedades del sistema circulatorio) lidera con una media de 50 horas, seguido del código 6 
(enfermedades del sistema nervioso) con 39.6 horas y el código 12 (enfermedades de la piel) 
con 39.5 horas. Estos tres motivos, aunque poco frecuentes, generan las ausencias más largas 
y costosas operativamente.

En el extremo opuesto, los motivos más frecuentes del dataset — código 23 (consulta médica, 
222 registros) y código 28 (ausencia injustificada, 158 registros) — presentan medias muy 
bajas de 2.8 y 3.0 horas respectivamente, lo que indica que la mayoría de las ausencias 
son cortas pero recurrentes.

Es destacable el código 19 (traumatismos y envenenamientos) con una media de 23 horas y 
67 registros, siendo el motivo con mayor combinación de frecuencia e impacto en horas, 
lo que lo convierte en un foco de atención prioritario para la gestión del absentismo.
