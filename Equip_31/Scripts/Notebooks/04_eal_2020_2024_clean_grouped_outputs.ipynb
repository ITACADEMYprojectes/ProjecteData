```python
def read_csv_eal(doc_path):
    df = pd.read_csv(doc_path, encoding="utf-8", header=None)

    df = df.map(
        lambda x: re.sub(r"\s+", " ", str(x)).strip()
        if pd.notna(x) else pd.NA
    )

    df = df.replace(["", "nan", "NaN"], pd.NA)

    return df
```

```python
def process_eal16_2020_wide(df, anio=2020):
    df = df.copy()

    # 1. Extraer tabla y pregunta
    values = df.stack().dropna().astype(str)

    titulo = values[
        values.str.match(r"^EAL-\d+[A-Za-z]?\.", na=False)
    ].iloc[0]

    tabla = re.match(r"^(EAL-\d+[A-Za-z]?)", titulo).group(1)

    pregunta = re.sub(
        r"^EAL-\d+[A-Za-z]?\.\s*",
        "",
        titulo
    )

    # 2. Nombres finales de las columnas numéricas
    columnas_valores = [
        "TOTAL",
        "DE 5 A 9 TRABAJADORES",
        "DE 10 A 49 TRABAJADORES",
        "DE 50 A 249 TRABAJADORES",
        "DE 250 A 499 TRABAJADORES",
        "MÁS DE 499 TRABAJADORES"
    ]

    # 3. Detectar columnas que contienen valores numéricos
    patron_numero = r"^\s*\d+(?:[,.]\d+)?\s*$"

    numeric_like = df.map(
        lambda x: bool(re.match(patron_numero, str(x)))
        if pd.notna(x) else False
    )

    value_cols = [
        col for col in df.columns[1:]
        if numeric_like[col].sum() > 0
    ]

    # En caso de que detecte más columnas, nos quedamos con las 6 principales
    if len(value_cols) != 6:
        counts = numeric_like.iloc[:, 1:].sum().sort_values(ascending=False)
        value_cols = sorted(counts[counts > 0].head(6).index.tolist())

    if len(value_cols) != 6:
        raise ValueError(
            f"Se esperaban 6 columnas numéricas, "
            f"pero se detectaron {len(value_cols)}: {value_cols}"
        )

    # 4. Crear dataframe base
    col_competencia = df.columns[0]

    df_out = df[[col_competencia] + value_cols].copy()
    df_out.columns = ["competencia"] + columnas_valores

    # 5. Convertir columnas numéricas
    for col in columnas_valores:
        df_out[col] = (
            df_out[col]
            .astype(str)
            .str.replace(",", ".", regex=False)
            .str.replace("%", "", regex=False)
            .str.strip()
        )

        df_out[col] = pd.to_numeric(df_out[col], errors="coerce")

    # 6. Detectar filas de datos
    is_data = df_out[columnas_valores].notna().sum(axis=1) > 0

    # 7. Quedarnos solo con filas con datos
    df_out = df_out[is_data].copy()

    # 8. Limpiar competencia
    df_out["competencia"] = (
        df_out["competencia"]
        .astype(str)
        .str.replace("\n", " ", regex=False)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

    # 9. Eliminar posibles filas residuales
    df_out = df_out[
        ~df_out["competencia"].str.match(
            r"^EAL-|^Año\s+\d{4}|^\(\d+\)|^Porcentaje",
            case=False,
            na=False
        )
    ].copy()

    # 10. Insertar metadatos
    df_out.insert(0, "anio", anio)
    df_out.insert(1, "tabla", tabla)
    df_out.insert(2, "pregunta", pregunta)

    # 11. Orden final
    df_out = df_out[
        [
            "anio",
            "tabla",
            "pregunta",
            "competencia"
        ] + columnas_valores
    ].reset_index(drop=True)

    return df_out
```

```python
def process_eal16a_wide(df, anio=2020):
    df = df.copy()

    # 1. Extraer tabla y pregunta
    values = df.stack().dropna().astype(str)

    titulo = values[
        values.str.match(r"^EAL-\d+[A-Za-z]?\.", na=False)
    ].iloc[0]

    tabla = re.match(r"^(EAL-\d+[A-Za-z]?)", titulo).group(1)

    pregunta = re.sub(
        r"^EAL-\d+[A-Za-z]?\.\s*",
        "",
        titulo
    )

    # 2. Nombres finales de las columnas numéricas
    # En 2020 aparecen en dos bloques de 5 columnas
    columnas_valores = [
        "TOTAL",
        "INDUSTRIA",
        "CONSTRUCCIÓN",
        "COMERCIO Y REPARACIÓN DE VEHÍCULOS",
        "TRANSPORTE Y ALMACENAMIENTO",
        "HOSTELERÍA",
        "INFORMACIÓN, COMUNICACIONES Y ACTIVIDADES FINANCIERAS",
        "ACTIVIDADES PROFESIONALES, CIENTÍFICAS Y TÉCNICAS",
        "ACTIVIDADES ADMINISTRATIVAS Y SERVICIOS AUXILIARES",
        "ACTIVIDADES ARTÍSTICAS, RECREATIVAS Y DE ENTRETENIMIENTO Y OTROS SERVICIOS"
    ]

    # 3. Detectar columnas que contienen valores numéricos
    patron_numero = r"^\s*\d+(?:[,.]\d+)?\s*$"

    numeric_like = df.map(
        lambda x: bool(re.match(patron_numero, str(x)))
        if pd.notna(x) else False
    )

    value_cols = [
        col for col in df.columns[1:]
        if numeric_like[col].sum() > 0
    ]

    if len(value_cols) == 0:
        raise ValueError("No se han detectado columnas numéricas en EAL-16a")

    # En esta tabla se detectan 5 columnas numéricas por bloque
    block_width = len(value_cols)

    if len(columnas_valores) % block_width != 0:
        raise ValueError(
            f"Las columnas esperadas ({len(columnas_valores)}) "
            f"no encajan con el ancho de bloque detectado ({block_width})"
        )

    expected_blocks = len(columnas_valores) // block_width

    # 4. Detectar filas con datos numéricos
    is_data = numeric_like[value_cols].sum(axis=1) > 0

    data_indices = df.index[is_data].tolist()

    # 5. Separar bloques de filas consecutivas
    data_groups = []
    current_group = []

    for idx in data_indices:
        if not current_group:
            current_group = [idx]
        elif idx == current_group[-1] + 1:
            current_group.append(idx)
        else:
            data_groups.append(current_group)
            current_group = [idx]

    if current_group:
        data_groups.append(current_group)

    if len(data_groups) < expected_blocks:
        raise ValueError(
            f"Se esperaban {expected_blocks} bloques de datos, "
            f"pero se detectaron {len(data_groups)}"
        )

    # Nos quedamos con los bloques necesarios
    data_groups = data_groups[:expected_blocks]

    # 6. Procesar cada bloque
    col_competencia = df.columns[0]
    blocks = []

    for block_idx, group_indices in enumerate(data_groups):

        start = block_idx * block_width
        end = start + block_width

        block_column_names = columnas_valores[start:end]

        df_block = df.loc[
            group_indices,
            [col_competencia] + value_cols
        ].copy()

        df_block.columns = ["competencia"] + block_column_names

        # Limpiar competencia
        df_block["competencia"] = (
            df_block["competencia"]
            .astype(str)
            .str.replace("\n", " ", regex=False)
            .str.replace(r"\s+", " ", regex=True)
            .str.strip()
        )

        # Convertir columnas numéricas
        for col in block_column_names:
            df_block[col] = (
                df_block[col]
                .astype(str)
                .str.replace(",", ".", regex=False)
                .str.replace("%", "", regex=False)
            )

            df_block[col] = pd.to_numeric(df_block[col], errors="coerce")

        # Mantener solo filas reales con datos
        df_block = df_block[
            df_block[block_column_names].notna().sum(axis=1) > 0
        ].copy()

        # Eliminar posibles filas residuales
        df_block = df_block[
            ~df_block["competencia"].str.match(
                r"^EAL-|^Año\s+\d{4}|^\(\d+\)|^Porcentaje",
                case=False,
                na=False
            )
        ].copy()

        df_block = df_block.reset_index(drop=True)

        blocks.append(df_block)

    # 7. Unir los bloques por competencia
    df_out = blocks[0]

    for block in blocks[1:]:
        df_out = df_out.merge(
            block,
            on="competencia",
            how="outer"
        )

    # 8. Insertar metadatos
    df_out.insert(0, "anio", anio)
    df_out.insert(1, "tabla", tabla)
    df_out.insert(2, "pregunta", pregunta)

    # 9. Orden final
    df_out = df_out[
        [
            "anio",
            "tabla",
            "pregunta",
            "competencia"
        ] + columnas_valores
    ].reset_index(drop=True)

    return df_out
```

```python
def process_eal17_wide(df, anio=2024):
    df = df.copy()

    # 1. Extraer tabla y pregunta
    values = df.stack().dropna().astype(str)

    titulo = values[
        values.str.match(r"^EAL-\d+[A-Za-z]?\.", na=False)
    ].iloc[0]

    tabla = re.match(r"^(EAL-\d+[A-Za-z]?)", titulo).group(1)

    pregunta = re.sub(
        r"^EAL-\d+[A-Za-z]?\.\s*",
        "",
        titulo
    )

    # 2. Nombres finales de las columnas numéricas
    columnas_valores = [
        "total_empresas",
        "empresas_formacion_total",
        "formacion_cursos_y_otros",
        "solo_cursos",
        "solo_otros_tipos_formacion",
        "empresas_no_formacion"
    ]

    # 3. Detectar columnas que contienen valores numéricos
    patron_numero = r"^\s*\d+(?:[,.]\d+)?\s*$"

    numeric_like = df.map(
        lambda x: bool(re.match(patron_numero, str(x)))
        if pd.notna(x) else False
    )

    value_cols = [
        col for col in df.columns[1:]
        if numeric_like[col].sum() > 0
    ]

    # En caso de que detecte más columnas, nos quedamos con las 6 principales
    if len(value_cols) != 6:
        counts = numeric_like.iloc[:, 1:].sum().sort_values(ascending=False)
        value_cols = sorted(counts[counts > 0].head(6).index.tolist())

    if len(value_cols) != 6:
        raise ValueError(f"Se esperaban 6 columnas numéricas, pero se detectaron {len(value_cols)}: {value_cols}")

    # 4. Crear dataframe base
    col_ambito = df.columns[0]

    df_out = df[[col_ambito] + value_cols].copy()
    df_out.columns = ["ambito"] + columnas_valores

    # 5. Convertir columnas numéricas
    for col in columnas_valores:
        df_out[col] = (
            df_out[col]
            .astype(str)
            .str.replace(",", ".", regex=False)
            .str.replace("%", "", regex=False)
        )

        df_out[col] = pd.to_numeric(df_out[col], errors="coerce")

    # 6. Detectar filas de datos y filas de bloque
    is_data = df_out[columnas_valores].notna().sum(axis=1) > 0

    is_section = (
        df_out["ambito"].notna()
        & ~is_data
        & ~df_out["ambito"].astype(str).str.match(
            r"^EAL-|^Año\s+\d{4}|\(\d+\)",
            case=False,
            na=False
        )
    )

    # 7. Crear columna dimension
    df_out["dimension"] = df_out["ambito"].where(is_section).ffill()

    # 8. Quedarnos solo con filas con datos
    df_out = df_out[is_data].copy()

    # La primera fila TOTAL no tiene bloque previo
    df_out["dimension"] = df_out["dimension"].fillna("TOTAL")

    df_out.loc[
        df_out["ambito"].astype(str).str.upper().eq("TOTAL"),
        "dimension"
    ] = "TOTAL"

    # 9. Insertar metadatos
    df_out.insert(0, "anio", anio)
    df_out.insert(1, "tabla", tabla)
    df_out.insert(2, "pregunta", pregunta)

    # 10. Orden final
    df_out = df_out[
        [
            "anio",
            "tabla",
            "pregunta",
            "dimension",
            "ambito"
        ] + columnas_valores
    ].reset_index(drop=True)

    return df_out
```

```python
def transformation(df, anio=2024, filas_debajo_eliminar=2):
    df = df.copy()

    col_texto = "ambito"
    patron_tabla = r"^(EAL-\d+[A-Za-z]?)"

    # Detectar código de tabla: EAL-25, EAL-25a, EAL-25b...
    tabla_detectada = (
        df[col_texto]
        .astype(str)
        .str.extract(patron_tabla, expand=False)
    )

    # Filas donde empieza una tabla
    mask_tabla = tabla_detectada.notna()

    # Pregunta completa asociada a cada bloque
    pregunta_detectada = (
        df[col_texto]
        .where(mask_tabla)
        .ffill()
        .astype(str)
        .str.replace(r"^EAL-\d+[A-Za-z]?\.\s*", "", regex=True)
    )

    # Insertar columnas al principio
    df.insert(0, "anio", anio)
    df.insert(1, "tabla", tabla_detectada.ffill())
    df.insert(2, "pregunta", pregunta_detectada)

    # Eliminar fila de pregunta + N filas inferiores
    mask_eliminar = pd.concat(
        [
            mask_tabla.shift(i, fill_value=False)
            for i in range(filas_debajo_eliminar + 1)
        ],
        axis=1
    ).any(axis=1)

    df = df.loc[~mask_eliminar].reset_index(drop=True)

    return df
```

```python
def process_ccaa_long(df, anio=2024):
    df = df.copy()

    # -----------------------------
    # 1. Extract table and question
    # -----------------------------
    values = df.stack().dropna().astype(str)

    titulo = values[
        values.str.match(r"^EAL-\d+[A-Za-z]?\.", na=False)
        ].iloc[0]

    tabla = re.match(r"^(EAL-\d+[A-Za-z]?)", titulo).group(1)

    pregunta = re.sub(
        r"^EAL-\d+[A-Za-z]?\.\s*",
        "",
        titulo
    )

    # -----------------------------
    # 2. Detect CCAA header rows
    # -----------------------------
    ccaa_keywords = [
        "TOTAL", "ANDALUCÍA", "ARAGÓN", "ASTURIAS (PRINCIPADO DE)", "BALEARES (ILLES)",
        "CANARIAS", "CANTABRIA", "CASTILLA-LA MANCHA", "CASTILLA Y LEÓN", "CATALUÑA",
        "COMUNITAT VALENCIANA", "EXTREMADURA", "GALICIA", "MADRID (COMUNIDAD DE)",
        "MURCIA (REGIÓN DE)", "NAVARRA (COMUNIDAD FORAL DE)", "PAÍS VASCO", "RIOJA, LA"
    ]

    patron_ccaa = "|".join(re.escape(x) for x in ccaa_keywords)

    mask_headers = df.apply(
        lambda row: row.astype(str)
        .str.contains(patron_ccaa, case=False, regex=True)
        .sum() >= 2,
        axis=1
    )

    header_rows = df.index[mask_headers].tolist()

    # -----------------------------
    # 3. Process each horizontal block
    # -----------------------------
    bloques = []
    col_competencia = df.columns[0]

    for i, header_row in enumerate(header_rows):

        start_row = header_row + 1

        if i + 1 < len(header_rows):
            end_row = header_rows[i + 1]
        else:
            end_row = len(df)

        header = df.loc[header_row]

        value_cols = header[header.notna()].index.tolist()
        value_cols = [c for c in value_cols if c != col_competencia]

        nombres_ccaa = header[value_cols].tolist()

        block = df.loc[start_row:end_row - 1, [col_competencia] + value_cols].copy()

        block.columns = ["competencia"] + nombres_ccaa

        block = block[block["competencia"].notna()].copy()

        block_long = block.melt(
            id_vars="competencia",
            var_name="comunidad_autonoma",
            value_name="porcentaje"
        )

        bloques.append(block_long)

    # -----------------------------
    # 4. Concatenate and clean values
    # -----------------------------
    df_final = pd.concat(bloques, ignore_index=True)

    df_final["porcentaje"] = (
        df_final["porcentaje"]
        .astype(str)
        .str.replace(",", ".", regex=False)
        .str.replace("%", "", regex=False)
    )

    df_final["porcentaje"] = pd.to_numeric(
        df_final["porcentaje"],
        errors="coerce"
    )

    df_final = df_final.dropna(subset=["porcentaje"]).reset_index(drop=True)

    # -----------------------------
    # 5. Add metadata
    # -----------------------------
    df_final.insert(0, "anio", anio)
    df_final.insert(1, "tabla", tabla)
    df_final.insert(2, "pregunta", pregunta)

    return df_final
```

```python
def to_long_format(
    df,
    id_vars,
    value_vars,
    var_name,
    value_name="porcentaje"
):
    df = df.copy()

    df_long = df.melt(
        id_vars=id_vars,
        value_vars=value_vars,
        var_name=var_name,
        value_name=value_name
    )

    df_long[value_name] = (
        df_long[value_name]
        .astype(str)
        .str.replace(",", ".", regex=False)
        .str.replace("%", "", regex=False)
    )

    df_long[value_name] = pd.to_numeric(
        df_long[value_name],
        errors="coerce"
    )

    df_long = df_long.dropna(subset=[value_name]).reset_index(drop=True)

    return df_long
```

```python
def parse_eal_number(x):
    if pd.isna(x):
        return pd.NA

    s = str(x).strip()

    if s == "" or s.lower() in {"nan", "<na>", "none"}:
        return pd.NA

    s = s.replace("%", "")
    s = re.sub(r"\s+", "", s)

    if not re.match(r"^-?[\d.,]+$", s):
        return pd.NA

    if "," in s and "." in s:
        # Ejemplo: 1.234,5
        s = s.replace(".", "").replace(",", ".")

    elif "," in s:
        # Ejemplo: 42,7
        s = s.replace(",", ".")

    elif "." in s:
        partes = s.split(".")

        if len(partes) > 2:
            # Ejemplo: 10.932.782
            s = "".join(partes)

        else:
            izquierda, derecha = partes

            if len(derecha) == 3 and len(izquierda) <= 3:
                # Ejemplo: 352.936 como separador de miles
                s = izquierda + derecha
            else:
                # Ejemplo: 42.7 como decimal
                s = izquierda + "." + derecha

    return pd.to_numeric(s, errors="coerce")


def is_eal_number(x):
    return pd.notna(parse_eal_number(x))


def detect_eal_value_cols(df, columnas_valores, tabla=None, anio=None):
    n_expected = len(columnas_valores)

    numeric_like = df.map(is_eal_number)
    numeric_counts = numeric_like.sum(axis=0)

    detected = [
        col for col in df.columns[1:]
        if numeric_counts[col] > 1
    ]

    cols_order = list(df.columns)

    detected = sorted(
        detected,
        key=lambda c: cols_order.index(c)
    )

    if len(detected) == n_expected:
        return detected

    # Caso típico:
    # detecta [1, 2, 3, 4, 5, 7]
    # pero la tabla realmente usa [1, 2, 3, 4, 5, 6, 7]
    if detected:
        start_pos = cols_order.index(detected[0])

        candidate = cols_order[start_pos:start_pos + n_expected]

        if len(candidate) == n_expected:
            return candidate

    raise ValueError(
        f"Se esperaban {n_expected} columnas numéricas, "
        f"pero se detectaron {len(detected)}: {detected}. "
        f"Conteo numérico por columna: {numeric_counts.to_dict()}. "
        f"Tabla: {tabla}, Año: {anio}"
    )
```

```python
def process_ealc1_wide(df, anio=None):
    df = df.copy()

    # 1. Extraer tabla, pregunta y año
    values = df.stack().dropna().astype(str)

    titulo = values[
        values.str.match(r"^EAL-[A-Za-z]?\d+[A-Za-z]?\.", na=False)
    ].iloc[0]

    tabla = re.match(
        r"^(EAL-[A-Za-z]?\d+[A-Za-z]?)",
        titulo
    ).group(1)

    pregunta = re.sub(
        r"^EAL-[A-Za-z]?\d+[A-Za-z]?\.\s*",
        "",
        titulo
    )

    # Eliminar nota final tipo (1), si existe
    pregunta = re.sub(r"\s*\(\d+\)\s*$", "", pregunta).strip()

    if anio is None:
        texto_anio = values[
            values.str.contains(r"Año\s+\d{4}", case=False, regex=True, na=False)
        ]

        if len(texto_anio) > 0:
            anio = int(re.search(r"\d{4}", texto_anio.iloc[0]).group(0))
        else:
            anio = pd.NA

    # 2. Nombres finales de columnas numéricas
    columnas_valores = [
        "total_empresas",
        "total_trabajadores",
        "varones",
        "mujeres",
        "distribucion_porcentual_varones",
        "distribucion_porcentual_mujeres"
    ]

    # 3. Detectar columnas con valores numéricos
    numeric_like = df.map(is_eal_number)

    value_cols = [
        col for col in df.columns[1:]
        if numeric_like[col].sum() > 1
    ]

    value_cols = sorted(value_cols)

    if len(value_cols) != len(columnas_valores):
        raise ValueError(
            f"Se esperaban {len(columnas_valores)} columnas numéricas, "
            f"pero se detectaron {len(value_cols)}: {value_cols}"
        )
    # 4. Detectar columna o columnas de ámbito
    first_value_pos = df.columns.get_loc(value_cols[0])
    label_cols = list(df.columns[:first_value_pos])

    df_base = df[label_cols + value_cols].copy()

    # Unir posibles columnas de texto antes de las numéricas
    df_base["ambito"] = df_base[label_cols].apply(
        lambda row: " ".join(
            row.dropna().astype(str).str.strip()
        ).strip(),
        axis=1
    )

    df_base["ambito"] = df_base["ambito"].replace("", pd.NA)

    df_out = df_base[["ambito"] + value_cols].copy()
    df_out.columns = ["ambito"] + columnas_valores

    # 5. Convertir columnas numéricas
    for col in columnas_valores:
        df_out[col] = df_out[col].map(parse_eal_number)

    # 6. Detectar filas con datos reales
    is_data = df_out[columnas_valores].notna().sum(axis=1) > 0

    # 7. Detectar bloques o dimensiones
    section_names = {
        "TAMAÑO DE LA EMPRESA",
        "ACTIVIDAD ECONÓMICA",
        "COMUNIDAD AUTÓNOMA"
    }

    ambito_norm = (
        df_out["ambito"]
        .fillna("")
        .astype(str)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
        .str.upper()
    )

    is_section = (
        ~is_data
        & ambito_norm.isin(section_names)
    )

    # 8. Ignorar filas de título, año, cabeceras y notas
    is_ignore = (
        ~is_data
        & ~is_section
        & df_out["ambito"].fillna("").astype(str).str.match(
            r"^EAL-|^Año\s+\d{4}|^Volver al índice|^\(\d+\)",
            case=False,
            na=False
        )
    )

    # 9. Reconstruir ámbitos partidos en varias filas, si los hubiera
    pending_text = []

    for idx in df_out.index:
        ambito = df_out.at[idx, "ambito"]

        if pd.isna(ambito):
            ambito_txt = ""
        else:
            ambito_txt = str(ambito).strip()

        if is_data.loc[idx]:
            if pending_text:
                textos = pending_text.copy()

                if ambito_txt:
                    textos.append(ambito_txt)

                df_out.at[idx, "ambito"] = " ".join(textos)
                pending_text = []

        elif is_section.loc[idx]:
            pending_text = []

        elif not is_ignore.loc[idx] and ambito_txt:
            pending_text.append(ambito_txt)

    # 10. Crear columna dimension
    df_out["dimension"] = df_out["ambito"].where(is_section).ffill()

    # 11. Quedarnos solo con filas de datos
    df_out = df_out[is_data].copy()

    df_out["dimension"] = df_out["dimension"].fillna("TOTAL")

    df_out.loc[
        df_out["ambito"].astype(str).str.upper().eq("TOTAL"),
        "dimension"
    ] = "TOTAL"

    # 12. Añadir metadatos
    df_out.insert(0, "anio", anio)
    df_out.insert(1, "tabla", tabla)
    df_out.insert(2, "pregunta", pregunta)

    # 13. Orden final
    df_out = df_out[
        [
            "anio",
            "tabla",
            "pregunta",
            "dimension",
            "ambito"
        ] + columnas_valores
    ].reset_index(drop=True)

    return df_out
```

# Procesamiento

### Formación

```python
# ============================================================
# 4. Procesamiento en memoria
# ============================================================

processed_dfs_formacion = {}
subset_dfs_formacion = {}

for table_name, config in TABLE_CONFIG.items():

    processed_dfs_formacion[table_name] = {}
    subset_dfs_formacion[table_name] = {}

    for year in YEARS:

        input_path = INPUT_BASE_FOLDER / year / f"{table_name}.csv"

        if not input_path.exists():
            print(f"No encontrado: {table_name} - {year}")
            continue

        # Funciones base
        reader_function = config["reader"]
        transform_function = config["transformer"]
        subset_function = config.get("subsetter")

        # Sobrescritura específica por año, si existe
        year_config = config.get("year_overrides", {}).get(year, {})

        reader_function = year_config.get("reader", reader_function)
        transform_function = year_config.get("transformer", transform_function)
        subset_function = year_config.get("subsetter", subset_function)

        # 1. Leer
        df_raw = reader_function(input_path)

        # 2. Transformar
        df_processed = transform_function(
            df_raw,
            anio=int(year)
        )

        # 3. Asegurar año
        if "anio" not in df_processed.columns:
            df_processed["anio"] = int(year)

        processed_dfs_formacion[table_name][year] = df_processed

        # 4. Subset
        if subset_function is not None:
            df_subset = subset_function(df_processed)
        else:
            df_subset = df_processed.copy()

        subset_dfs_formacion[table_name][year] = df_subset

        print(
            f"Procesado: {table_name} - {year} | "
            f"reader: {reader_function.__name__} | "
            f"transformer: {transform_function.__name__} | "
            f"processed: {df_processed.shape} | "
            f"subset: {df_subset.shape}"
        )
```

### Contexto y muestreo

```python
# ============================================================
# 4. Procesamiento en memoria
# ============================================================

processed_dfs_contexto_muestreo = {}
subset_dfs_contexto_muestreo = {}

for table_name, config in TABLE_CONFIG_CONTEXTO_MUESTREO.items():

    processed_dfs_contexto_muestreo[table_name] = {}
    subset_dfs_contexto_muestreo[table_name] = {}

    for year in YEARS:

        input_path = INPUT_BASE_FOLDER / year / f"{table_name}.csv"

        if not input_path.exists():
            print(f"No encontrado: {table_name} - {year}")
            continue

        # Funciones base
        reader_function = config["reader"]
        transform_function = config["transformer"]
        subset_function = config.get("subsetter")

        # Sobrescritura específica por año, si existe
        year_config = config.get("year_overrides", {}).get(year, {})

        reader_function = year_config.get("reader", reader_function)
        transform_function = year_config.get("transformer", transform_function)
        subset_function = year_config.get("subsetter", subset_function)

        # 1. Leer
        df_raw = reader_function(input_path)

        # 2. Transformar
        df_processed = transform_function(
            df_raw,
            anio=int(year)
        )

        # 3. Asegurar año
        if "anio" not in df_processed.columns:
            df_processed["anio"] = int(year)

        processed_dfs_contexto_muestreo[table_name][year] = df_processed

        # 4. Subset
        if subset_function is not None:
            df_subset = subset_function(df_processed)
        else:
            df_subset = df_processed.copy()

        subset_dfs_contexto_muestreo[table_name][year] = df_subset

        print(
            f"Procesado: {table_name} - {year} | "
            f"reader: {reader_function.__name__} | "
            f"transformer: {transform_function.__name__} | "
            f"processed: {df_processed.shape} | "
            f"subset: {df_subset.shape}"
        )
```

```python
# ============================================================
# 5. Combinar diccionarios y concatenar todos los años por tabla
# ============================================================

# 1. Combinar los dos diccionarios de subsets
subset_dfs_all = {
    **subset_dfs_formacion,
    **subset_dfs_contexto_muestreo
}

# 2. Concatenar todos los años por tabla
combined_subsets = {}

for table_name, year_dfs in subset_dfs_all.items():

    if year_dfs:
        combined_subsets[table_name] = pd.concat(
            year_dfs.values(),
            ignore_index=True
        )

        print(
            f"{table_name} concatenado: "
            f"{combined_subsets[table_name].shape}"
        )

    else:
        print(f"{table_name} no tiene datos para concatenar")
```
