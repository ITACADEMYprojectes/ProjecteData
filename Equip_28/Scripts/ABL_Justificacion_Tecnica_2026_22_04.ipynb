# Justificación Técnica

# Técnica Elegida 
Analisis descriptiva con segmentación temporal

# **1) ¿Por qué elegiste esta técnica de análisis y no otra?**  

Usamos un análisis descriptivo con segmentación temporal porque el objetivo era medir la magnitud del ausentismo y entender cómo varía a lo largo del año.
Con el dataset ya limpio —con nulos controlado, sin errores de tipado y con variables temporales correctas— esta técnica permite:
- cuantificar horas de absentismo por mes y estación
- identificar picos operativos y patrones estacionales
- analizar perfiles de riesgo mediante correlaciones simples
Es la opción más adecuada porque los datos son transversales y no permiten modelos temporales avanzados.

---

# **2) ¿Qué supuestos tiene la técnica que usaste y cómo verificaste que tus datos los cumplían?**  

Verificamos:
- Integridad del dataset: tipos corregidos, duplicados eliminados, variables binarias normalizadas.
- Independencia: cada fila representa un empleado distinto.
- Distribución: el absentismo es right‑skewed, por lo que usamos mediana y percentiles además de la media.
- Correlación: solo incluimos variables numéricas y comprobamos linealidad aproximada.
- Tamaño de muestra: 740 registros, suficiente para análisis descriptivo robusto

---

# **3) ¿Qué limitaciones tiene esta técnica y cómo podrían afectar a tus conclusiones?**  

- Los datos son transversales, no longitudinales: no permiten analizar recurrencia mensual por empleado ni predecir absentismo futuro.
- Los outliers extremos inflan la media y pueden distorsionar la magnitud del absentismo.
- La técnica no permite causalidad, solo asociaciones.
Impacto:
Podemos identificar meses críticos y perfiles de riesgo, pero no explicar causas profundas ni anticipar absentismo individual.

Mejoras futuras:
- incorporar datos longitudinales (si datos disponibles)
- aplicar modelos predictivos (datos historicos, variables explicativas)
- integrar variables de salud, clima laboral y carga real de trabajo

El desarrollo de modelos predictivos exige contar con datos históricos adicionales que hoy no están disponibles.
