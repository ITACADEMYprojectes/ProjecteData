```python
import pandas as pd

input_path = r"data/staySpain_cleaned.pkl"
df = pd.read_pickle(input_path)



columns = ["availability_30", "availability_60", "availability_90", "availability_365"]

avg_availability = (
    df.groupby("city", dropna=False)[columns]
      .mean(numeric_only=True)
      .round(1)
      .sort_index()
)
avg_availability
```
