## análisis de grupo con los 3 ptos

```python
# este df son solo los clientes que tienen todos los 3 pctos 

# Selección de columnas
columnas = ['id', 'age_group', 'job', 'marital', 'education', 'housing', 'loan', 'deposit']

# Filtro: solo filas donde housing, loan y deposit son 'yes'
filtro = (df['housing'] == 'yes') & (df['loan'] == 'yes') & (df['deposit'] == 'yes')

# Aplicar filtro y seleccionar columnas

demog_financieros = df.loc[filtro, columnas]

demog_financieros.head()
```

```python
# número de clientes con los 3 productos contratados
len(demog_financieros)
```

```python
# % de los clientes con los 3 productos contratados sobre el total de los clientes
porcentaje_demog_financieros = round((len(demog_financieros) / len(df))*100, 2)
porcentaje_demog_financieros
```

```python
# % de grupos etarios de los clientes con 3 ptos contratados
age_pct = (demog_financieros['age_group'].value_counts(normalize=True) * 100).round(2)
age_pct
```

```python
# gráfico de barras de % de estado civil de los clientes con 3 ptos contratados
marital_pct.plot.bar()
```

```python
# gráfico de barras de % de nivel educativo de los clientes con 3 ptos contratados
education_pct.plot.bar()
```

```python
# definir las variables demograficas para analizar
caract_demogr = ['age_group', 'marital', 'education', 'job']

# crear lista vacía para añadir resultado de la comparación
lista_comparacion = []

# iterar por cada variable demográfica para calcular % en 3ptos vs % de total clientes
for var in caract_demogr:
    # % para 3 ptos contratados
    subset_pct = demog_financieros[var].value_counts(normalize=True) * 100

    # % del total de los clientes
    total_pct = df[var].value_counts(normalize=True) * 100

    # combinar y calcular diferencia
    comparar = pd.DataFrame({'3products_%': subset_pct.round(2), 'total_%': total_pct.round(2)})

    comparar['difference'] = (comparar['3products_%'] - comparar['total_%']).round(2)
    comparar['variable'] = var

    # añadir en lista_comparacion
    lista_comparacion.append(comparar)

# concatenar todo en una tabla, renombrar columna index a category y ordenar valores
comparar_resumen = pd.concat(lista_comparacion).reset_index().rename(columns={'index': 'category'}).sort_values(['variable', 'difference'], ascending=[True, False])

comparar_resumen
```

```python
# graficar el resumen anterior de comparación por características demográficas y % de clientes con los 3ptos vs su % sobre el total de clientes

plt.figure(figsize=(10,6))
sns.barplot(data=comparar_resumen, x='difference', y='category', hue='variable', dodge=False)
plt.title('Comparación de los clientes con 3 productos contratados')
plt.xlabel('Diferencia vs Clientes Total (%)')
plt.ylabel('')
plt.axvline(0, color='gray', linestyle='--')
plt.tight_layout()
plt.show()
```
