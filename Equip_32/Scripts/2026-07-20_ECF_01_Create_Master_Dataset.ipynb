```python
# ======================================================
# DICCIONARIO DE BLOQUES
# ======================================================

VARIABLE_BLOCKS = {

    "COMMON": COMMON_VARS,

    "BEHAVIOUR": FINANCIAL_BEHAVIOUR,

    "ATTITUDES": ATTITUDES,

    "HOUSING": HOUSING,

    "FRAGILITY": FRAGILITY,

    "BACKGROUND": BACKGROUND,

    "COMPETENCIES": FINANCIAL_COMPETENCIES

}
```

```python
# ======================================================
# CREACIÓN DEL MASTER DATASET
# ======================================================

MASTER_VARIABLES = []

for block in VARIABLE_BLOCKS.values():

    MASTER_VARIABLES.extend(block)

MASTER_VARIABLES = list(dict.fromkeys(MASTER_VARIABLES))

print(f"Variables seleccionadas: {len(MASTER_VARIABLES)}")
```

```python
master_df = df[MASTER_VARIABLES].copy()
```

```python
# ======================================================
# COMPROBACIONES
# ======================================================

print("="*70)

print(f"Observaciones : {master_df.shape[0]:,}")

print(f"Variables     : {master_df.shape[1]}")

print("="*70)
```

```python
missing = (
    master_df
    .isna()
    .sum()
    .sort_values(ascending=False)
)

display(missing.head(30))
```

```python
# ======================================================
# EXPORTACIÓN
# ======================================================

master_df.to_csv(

    "../Data/2026-07-20_ECF_2021_MasterDataset.csv",

    index=False

)

print("✅ Dataset exportado")
```
