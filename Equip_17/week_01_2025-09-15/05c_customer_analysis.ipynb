# Quina és la puntuació mitjana donada pels usuaris als allotjaments turístics?

```python
avg_rating = df["review_scores_rating"].mean().round(2)

print(f"La puntuació mitjana donada pels usuaris als allotjaments turístics és de {avg_rating}")
```

# Quin percentatge d'allotjaments tenen una valoració general més gran de 80 a cada ciutat?

```python
import pandas as pd

# Creamos una columna nueva que será TRUE si el review_scores_rating es superior a 80
df["rating_greather_than_80"] = df["review_scores_rating"] > 80

df.head()
```

```python
# Agrupamos por ciudad y calculamos el porcentaje
df_grouped = (df.groupby("city")["rating_greather_than_80"].mean() * 100).round(2).reset_index()

df_grouped
```

```python
# Ordenamos por porcentaje en orden desc
df_sorted = df_grouped.sort_values(by="rating_greather_than_80", ascending=False).reset_index(drop=True)
```

```python
# Renombramos la columna
df_sorted = df_sorted.rename(columns={"rating_greather_than_80": "%_rating_greather_than_80"})

df_sorted
```
