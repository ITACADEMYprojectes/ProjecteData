```python
import pandas as pd

df = pd.read_pickle(r"data\staySpain_cleaned.pkl")

columns = ["availability_30", "availability_60", "availability_90", "availability_365"]

avg_availability = (
    df.groupby("city", dropna=False)[columns]
      .mean(numeric_only=True)
      .round(1)
      .sort_index()
)
avg_availability
```
