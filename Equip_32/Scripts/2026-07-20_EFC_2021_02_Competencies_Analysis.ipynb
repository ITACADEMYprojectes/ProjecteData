## 1. Tipificación de variables

Dentro del bloque de Competencias hay varios tipos de variable que requieren tratamientos distintos antes de poder construir cualquier índice o gráfico:

| Tipo | Variables | Tratamiento previsto |
|---|---|---|
| Escala Likert 1-5 | `d0101`-`d0117` | Ordinal, mantener escala |
| Escala 1-5 + categoría especial 6 | `c0100`, `c0500` | Ordinal + flag binario "no planifica" |
| Numéricas abiertas | `e0500`, `e0700`, `e0800`, `e0900` | Puntuar correcto/incorrecto contra el valor esperado |
| Verdadero/Falso | `e1001`, `e1002`, `e1003`, `e1200` | Recodificar a 1/0 correcto |
| Multi-respuesta (dummies) | `c0200a-k`, `c0600a-k`, `a1420a-g`, `d0610a-e` | Ya vienen como Sí=1/No=0 por variable |
| Categórica nominal | `f1100x` | Persona de referencia dentro del hogar |

```python
# Códigos especiales presentes en (casi) todas las variables de la ECF:
# -97 = No sabe | -98 = No ha lugar (no aplica el filtro) | -99 = No contesta
CODIGOS_ESPECIALES = [-97, -98, -99]


def limpiar_codigos_especiales(df, columnas, codigos=CODIGOS_ESPECIALES):
    """Sustituye los códigos especiales de la ECF por NaN en las columnas indicadas."""
    df_limpio = df.copy()
    for col in columnas:
        df_limpio[col] = df_limpio[col].replace(codigos, pd.NA)
    return df_limpio
```
