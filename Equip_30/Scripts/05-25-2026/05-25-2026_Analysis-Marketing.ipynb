# Pregunta:

Qué relación hay entre el número de contactos realizados durante esta campaña y la tasa de éxito, y como podemos optimizar la frecuencia de contacto para maximizar los resultados de nuestras campañas de marketing?



```python
# Calcular correlación de Spearman
corr, p_valor = spearmanr(df_marketing["campaign"], df_marketing["deposit"])

print(f"Correlación de Spearman: {corr:.4f}")
print(f"P-valor: {p_valor:.4f}")


#Gráfico
fig = px.scatter(
    df_campaign,
    x="campaign",
    y="tasa_éxito",
    size="n_clientes",        # tamaño del punto proporcional a clientes
    size_max=60,  # tamaño máximo de burbuja
    trendline="lowess",
    title=f"Número de contactos por tasa de éxito [ Correlación de Spearman: {corr:.4f} ] ",
    labels={"campaign": "Número de contactos", "tasa_éxito": "tasa de éxito"}
)

fig.show()
```

Interpretación:

Correlación -0.1287: relación negativa débil. A más contactos, menor tasa de éxito.


P-valor 0.0000: la relación es estadísticamente significativa, no es azar.

"Existe una relación negativa débil pero estadísticamente confirmada entre el número de contactos y la conversión."

# ¿Por qué Correlación Spearman en este análisis?

campaign tiene tres problemas que invalidan Pearson:

No es normal: la mayoría de valores se concentran en 1-5 con una cola larga hacia valores extremos (hasta 63).

Tiene outliers significativos:  ya confirmados visualmente en los gráficos anteriores. Pearson es muy sensible a estos y los resultados quedarían distorsionados.

La relación no tiene por qué ser lineal:  no esperamos que cada contacto adicional reduzca la conversión en una cantidad fija. Spearman detecta cualquier tendencia monótona.

# Número óptimo

```python
# Cálculo Número Óptimo  /// ## MEDIA / PROMEDIO

df_optimo = (
    df_marketing.groupby("campaign")
    .agg(
        tasa_éxito=("deposit", "mean"), ## MEDIA / PROMEDIO
        duracion_promedio=("duration", "mean"), ## MEDIA / PROMEDIO
        n_clientes=("deposit", "count") ## MEDIA / PROMEDIO
    )
    .reset_index()
)

df_optimo["tasa_éxito"] = df_optimo["tasa_éxito"] * 100
df_optimo["indice_optimo"] = df_optimo["tasa_éxito"] * df_optimo["duracion_promedio"]

# Filtro de representatividad
df_optimo_filtrado = df_optimo[df_optimo["n_clientes"] >= 10] ## fitrar más de 10 clientes

#  tasa de éxito + duración
opt = df_optimo_filtrado.loc[df_optimo_filtrado["indice_optimo"].idxmax()] #### busca la fila con la tasa de éxito más alta
print("\n--- Óptimo: tasa de éxito + duración  + clientes ---")
print(f"Contactos: {int(opt['campaign'])} | Tasa: {opt['tasa_éxito']:.2f}% | Duración: {opt['duracion_promedio']:.0f}s | Clientes: {int(opt['n_clientes'])}")
```

# Intervalo de confianza

Con n_clientes bajos, la media es poco fiable. 
Calcular error da credibilidad estadística a la conclusión del óptimo en 1 contacto.

--- intervalo de confianza al 95% // cálculo de error da credibilidad estadística
Sin error, una barra al 80% con 12 clientes parece igual de fiable que una al 80% con 500 clientes. Con el intervalo de confianza, la primera tendrá una barra de error enorme y la segunda casi imperceptible. El gráfico comunica visualmente dónde confiar en el dato y dónde no.


```python
## Intervalo de confianza

# Intervalo de confianza 95% sobre la tasa de éxito: cuanto menor es n_clientes, mayor es la barra de error

#df 
df_campaign["error"] = 1.96 * np.sqrt(  ## 1.96 = escala al 95% de confianza ## sqrt error estándar
    (df_campaign["tasa_éxito"] / 100) * (1 - df_campaign["tasa_éxito"] / 100) / df_campaign["n_clientes"]
) * 100

#gráfico
fig = px.bar(
    df_campaign,
    x="campaign", y="tasa_éxito",
    error_y="error",
    title="Tasa de éxito con intervalo de confianza 95%",
    labels={"campaign": "Número de contactos", "tasa_éxito": "Tasa de éxito (%)"}
)
fig.update_yaxes(ticksuffix="%")
fig.show()
```

Intervalo de confianza 95% sobre la tasa de éxito: cuanto menor es n_clientes, mayor es la barra de error

# Análisis efectividad y punto de saturación

```python
# GRÁFICO 1: Distribución de clientes por rangos (buckets) y tasa de conversión
# =============================================================

# Crear columna de rangos sobre df_marketing
bins = [0, 3, 5, 10, df_marketing["campaign"].max()]
labels = ["1 - 3 calls", "4 - 5 calls", "6 - 10 calls", "More than 10 calls"]
df_marketing["campaign_bucket"] = pd.cut(df_marketing["campaign"], bins=bins, labels=labels)

# Calcular métricas por bucket
df_buckets = (
    df_marketing.groupby("campaign_bucket", observed=True)
    .agg(
        n_clientes=("deposit", "count"),
        tasa_exito=("deposit", "mean")
    )
    .reset_index()
)
df_buckets["tasa_exito"] = df_buckets["tasa_exito"] * 100
df_buckets["pct_total"] = df_buckets["n_clientes"] / df_buckets["n_clientes"].sum() * 100

fig1 = make_subplots(specs=[[{"secondary_y": True}]])

fig1.add_trace(
    go.Bar(
        x=df_buckets["campaign_bucket"],
        y=df_buckets["n_clientes"],
        name="Clientes",
    ),
    secondary_y=False
)

fig1.add_trace(
    go.Scatter(
        x=df_buckets["campaign_bucket"],
        y=df_buckets["tasa_exito"],
        name="Tasa de conversión (%)",
        mode="lines+markers",
        line=dict(color="coral", width=2),
        marker=dict(size=8)
    ),
    secondary_y=True
)

for _, row in df_buckets.iterrows():
    fig1.add_annotation(
        x=row["campaign_bucket"],
        y=row["n_clientes"],
        text=f"{row['pct_total']:.2f}%",
        showarrow=False,
        yshift=10,
        font=dict(size=11)
    )

fig1.update_layout(
    title="Distribución de clientes y éxito en la contratación de depósitos",
    xaxis_title="Rangos de cantidad de contactos",
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
)
fig1.update_yaxes(title_text="Nº de clientes", secondary_y=False)
fig1.update_yaxes(title_text="Tasa de Conversión (%)", ticksuffix="%", secondary_y=True)
#fig.update_layout(title_subtitle="Distribución de clientes (% del total) y éxito en la contratación de depósitos por rangos de contacto (Buckets)")
fig1.show()
```

```python
# GRÁFICO 2: Efectividad y volumen por número individual de llamadas
# =============================================================

# Filtrar solo llamadas con >= 100 clientes (representatividad estadística)
df_campaign_100 = df_campaign[df_campaign["n_clientes"] >= 100].copy()

fig2 = make_subplots(specs=[[{"secondary_y": True}]])

fig2.add_trace(
    go.Bar(
        x=df_campaign_100["campaign"],
        y=df_campaign_100["n_clientes"],
        name="Clientes",
    ),
    secondary_y=False
)

fig2.add_trace(
    go.Scatter(
        x=df_campaign_100["campaign"],
        y=df_campaign_100["tasa_éxito"],
        name="Tasa de conversión (%)",
        mode="lines+markers",
        line=dict(color="coral", width=2),
        marker=dict(size=8)
    ),
    secondary_y=True
)

for _, row in df_campaign_100.iterrows():
    pct = row["n_clientes"] / df_campaign["n_clientes"].sum() * 100
    fig2.add_annotation(
        x=row["campaign"],
        y=row["n_clientes"],
        text=f"{pct:.2f}%",
        showarrow=False,
        yshift=10,
        font=dict(size=10)
    )

fig2.update_layout(
    title="Efectividad de la campaña y volumen de clientes por número de llamadas",
    xaxis_title="Número de Llamadas Realizadas",
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
)
fig2.update_yaxes(title_text="Nº de Clientes", secondary_y=False)
fig2.update_yaxes(title_text="Tasa de Conversión (%)", ticksuffix="%", secondary_y=True)
fig2.show()
```

```python
# GRÁFICO 3: Punto de saturación — dispersión tasa de conversión vs. llamadas
# =============================================================

fig3 = go.Figure()

fig3.add_trace(go.Scatter(
    x=df_campaign_100["campaign"],
    y=df_campaign_100["tasa_éxito"],
    mode="markers+text",
    text=df_campaign_100["campaign"].astype(str),
    textposition="top center",
    marker=dict(
        size=df_campaign_100["n_clientes"] / df_campaign_100["n_clientes"].max() * 40 + 8,
        color=df_campaign_100["tasa_éxito"],
        colorscale=[[0, "#2E4057"], [0.5, "#E8C99A"], [1, "#F4A261"]],
        showscale=False
    ),
    name="Llamadas"
))

fig3.add_hline(
    y=50,
    line_dash="dash",
    line_color="aquamarine",
    annotation_text="50% — límite de control",
    annotation_position="right"
)

fig3.update_layout(
    title="Punto de saturación: El éxito cae al aumentar la insistencia",
    xaxis_title="Número de llamadas realizadas",
    yaxis_title="Tasa de Conversión (%)",
    yaxis=dict(ticksuffix="%")
)
fig3.show()
```

```python
# GRÁFICO 4: Optimización de llamadas — real vs. optimizado por buckets
# =============================================================

# Simular escenario optimizado: máximo 3 contactos por cliente
df_marketing["campaign_capped"] = df_marketing["campaign"].clip(upper=3)

df_opt = (
    df_marketing.groupby("campaign_bucket", observed=True)
    .agg(
        actual=("campaign", "sum"),
        optimizado=("campaign_capped", "sum")
    )
    .reset_index()
)

fig4 = go.Figure()

fig4.add_trace(go.Bar(
    x=df_opt["campaign_bucket"],
    y=df_opt["actual"],
    name="Total de llamadas",
    text=df_opt["actual"],
    #marker_color="#E05C3A"
))

fig4.add_trace(go.Bar(
    x=df_opt["campaign_bucket"],
    y=df_opt["optimizado"],
    name="Total optimizado",
    text= df_opt["optimizado"], 
    marker_color="#F4A261",
    opacity=0.85
))

fig4.update_layout(
    title="Optimización de llamadas: Reducción del esfuerzo comercial innecesario",
    xaxis_title="Rangos de cantidad de contactos",
    yaxis_title="Total de llamadas",
    barmode="group",
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
)
fig4.show()
```
