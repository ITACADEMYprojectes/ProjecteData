# 2020

```python
import pandas as pd

df_25_2020 = pd.read_excel(
    r"C:\Users\crist\Documents\SIMULADOR\SPRINT4\TABLAS_EAL_2020.xlsx",
    sheet_name="EAL-25"
)

df_25_2020
```

## BARRERAS EMPRESAS + 50 TRABAJADORES (2020)

```python
# BLOQUE 50+ (filas aprox 37–46 según tu Excel)

df_50_2020 = df_25_2020.iloc[37:46, :6].copy()

df_50_2020.columns = ["Motivo", "Total", "Nada", "Poco", "Bastante", "Mucho"]

# limpiar filas raras
df_50_2020 = df_50_2020[df_50_2020["Motivo"].notna()]
df_50_2020 = df_50_2020[~df_50_2020["Motivo"].astype(str).str.startswith("(")]

# convertir a numérico
cols = ["Total", "Nada", "Poco", "Bastante", "Mucho"]
df_50_2020[cols] = df_50_2020[cols].astype(float)

# añadir info
df_50_2020["Grupo"] = "50+"
df_50_2020["Sector"] = "General"
df_50_2020["Año"] = 2020

df_50_2020
```

## BARRERAS SECTOR SERVICIOS (2020)

```python
# BLOQUE SERVICIOS (filas aprox 82–91)

df_servicios = df_25_2020.iloc[82:91, :6].copy()

df_servicios.columns = ["Motivo", "Total", "Nada", "Poco", "Bastante", "Mucho"]

# limpiar
df_servicios = df_servicios[df_servicios["Motivo"].notna()]
df_servicios = df_servicios[~df_servicios["Motivo"].astype(str).str.startswith("(")]

# convertir a numérico
cols = ["Total", "Nada", "Poco", "Bastante", "Mucho"]
df_servicios[cols] = df_servicios[cols].astype(float)

# añadir info
df_servicios["Grupo"] = "Todos"
df_servicios["Sector"] = "Servicios"
df_servicios["Año"] = 2020

df_servicios
```

## CONCAT (2020)

```python
df_final = pd.concat([df_50_2020, df_servicios], ignore_index=True)
df_final
```
