# Step 7: Handle Missing Values in Categorical Columns (`fillna` and `mode`)

In this step, missing values in the `education` and `marital` columns are replaced with the **most frequent value** within their respective groups (`job` for education, `age` for marital).

- Ensures that there are no NaN values in these categorical columns.
- Maintains logical consistency by filling values based on relevant groupings.
- This is a common imputation technique for categorical variables to improve data quality for analysis and machine learning.

```python
mode_per_job = (
    df1.groupby('job')['education']
        .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else None)
)

df1['education'] = df1['education'].fillna(df1['job'].map(mode_per_job))
```

```python
# Step 1: Find the most common marital status for each age
mode_per_age = (
    df1.groupby('age')['marital']
        .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else None)
)

# Step 2: Fill missing marital values based on age
df1['marital'] = df1['marital'].fillna(df1['age'].map(mode_per_age))
```

# Step 8: Handle Missing Values in Numeric Columns

In this step, missing values in the numeric column `age` are replaced with the **mean value** of the column.

- Ensures that there are no NaN values in numeric columns.  
- Maintains numeric type for analysis and machine learning.  
- This is a common imputation technique for continuous variables.

```python
# ------------------------------------------------------------
# Step 8: Fill Missing Values in Numeric Columns
# ------------------------------------------------------------
# Replace NaN values in the 'age' column with the column mean
# This keeps the data numeric and suitable for analysis or ML models
df1['age'] = df1['age'].fillna(df1['age'].mean())
```

# Step 13: Create a Complete Date Column and Identify the Day of the Week

In this step, we combine the *day*, *month*, and assigned *year* to create a full date for each record.  
This helps represent the data in a real calendar format, making it easier to analyze time-based patterns.  

We also extract the **day of the week** (e.g., Monday, Tuesday), which can provide useful insights into customer behavior based on the day of contact.

```python
month_map = {
    'jan': 1, 'feb': 2, 'mar': 3, 'apr': 4,
    'may': 5, 'jun': 6, 'jul': 7, 'aug': 8,
    'sep': 9, 'oct': 10, 'nov': 11, 'dec': 12
}
df1['month_num'] = df1['month'].str.lower().map(month_map)


start_year = 2008
years = []
current_year = start_year
for i, month in enumerate(df1['month_num']):
    if i == 0:
        current_year = start_year
    else:
        if month < df1['month_num'][i-1]:
            current_year += 1
        # Clip year to max 2010
        if current_year > 2010:
            current_year = 2010
    years.append(current_year)

df1['year'] = years


df1['date'] = pd.to_datetime({
    'year': df1['year'],
    'month': df1['month_num'],
    'day': df1['day']
}, errors='coerce')  # Invalid dates become NaT


df1['weekday'] = df1['date'].dt.day_name()


df1.drop(columns=['month_num'], inplace=True)
```
