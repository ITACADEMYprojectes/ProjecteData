```python
import pandas as pd
import numpy as np

# Replace df with your actual dataframe
import pandas as pd
df = pd.read_csv(r"C:\Personal\Susi\Bank Simulator\ECF files\09_Gestion_Gastos_y_Deudas_2016_GEN.csv", sep=";")
import pandas as pd
import numpy as np

# Replace system-missing codes
#df.replace([-97, -98, -99, -5], pd.NA, inplace=True)

# Rename columns for easier reference
df = df.rename(columns={
    # Saving
    "b1000a: ha estado los ultimos 12 meses: a.ahorrando dinero en metalico": "b1000a",
    "b1000b: ha estado los ultimos 12 meses: b.ahorrando en cuentas corrientes": "b1000b",
    "b1000c: ha estado los ultimos 12 meses: c.ahorrando en cuentas ahorro a plazo": "b1000c",
    "b1000d: ha estado los ultimos 12 meses: d.dando dinero a familiares para que...": "b1000d",
    "b1000e: ha estado los ultimos 12 meses: e.adquiriendo propiedades inmobiliarias.": "b1000e",
    "b1000f: ha estado los ultimos 12 meses: f.enviando remesas a familiares": "b1000f",
    "b1000g: ha estado los ultimos 12 meses: g.contribuyendo fondo de inversion...": "b1000g",
    "b1000h: ha estado los ultimos 12 meses: h.contributing personal pension": "b1000h",
    "b1000i: ha estado los ultimos 12 meses: i.aportando a su plan de pensiones": "b1000i",

    # Spending coping 
    "j0300a: gastos exceden ingresos? a.utilizar ahorros": "j0300a",
    "j0300b: gastos exceden ingresos? b.vender activos": "j0300b",
    "j0300c: gastos exceden ingresos? c.pedir prestado a familiares o a amigos": "j0300c",
    "j0300d: gastos exceden ingresos? d.pedir un adelanto...": "j0300d",
    "j0300e: gastos exceden ingresos? e.hipotecar un inmueble": "j0300e",
    "j0300f: gastos exceden ingresos? f.ampliar una hipoteca ya existente": "j0300f",
    "j0300g: gastos exceden ingresos? g.solicitar rescate anticipado plan pensiones": "j0300g",
    "j0300h: gastos exceden ingresos? h.empeniar algo de su propiedad": "j0300h",
    "j0300i: gastos exceden ingresos? i.disponer de una linea de credito existente...": "j0300i",
    "j0300j: gastos exceden ingresos? j.utilizar la tarjeta de credito...": "j0300j",
    "j0300k: gastos exceden ingresos? k.pedir un prestamo personal...": "j0300k",
    "j0300l: gastos exceden ingresos? l.pedir un prestamo de un proveedor...": "j0300l",
    "j0300m: gastos exceden ingresos? m.hacer un uso no autorizado del descubierto": "j0300m",
    "j0300n: gastos exceden ingresos? n.pagar con retraso o no pagar recibos": "j0300n",
    "j0300o: gastos exceden ingresos? o.otros": "j0300o",

    # Loan access
    "j0900a: ha experimentado..?: a.solicitaron prestamo que fue rechazado": "j0900a",
    "j0900b: ha experimentado..?: b.solicitaron prestamo y la cantidad concedida fue.": "j0900b",
    "j0900c: ha experimentado..?: c.no pidieron prestamo porque no se lo concederian.": "j0900c"
})

# -----------------------------
# Table 7 - Saving
saving_cols = ["b1000a","b1000b","b1000c","b1000d","b1000e","b1000f","b1000g","b1000h","b1000i"]

df["ahorra"] = (df[saving_cols] == 1).any(axis=1).astype(int)
df.loc[df[saving_cols].lt(0).any(axis=1), "ahorra"] = pd.NA

# -----------------------------
# Table 8 - Coping mechanisms
df["ahorros"] = ((df["j0300a"]==1) | (df["j0300b"]==1) | (df["j0300g"]==1)).astype(int)

df["cred_inf"] = ((df["j0300c"]==1) | (df["j0300d"]==1)).astype(int)

df["cred_line"] = ((df["j0300e"]==1) | (df["j0300f"]==1) | 
                   (df["j0300i"]==1) | (df["j0300j"]==1)).astype(int)

df["cred_add"] = ((df["j0300k"]==1) | (df["j0300l"]==1) | 
                  (df["j0300h"]==1)).astype(int)

df["impago"] = ((df["j0300m"]==1) | (df["j0300n"]==1)).astype(int)

df["ayudas"] = (df["j0300o"]==1).astype(int)

# -----------------------------
# Table 9 - Loan access
df["rechazo"] = (df["j0900a"] == 1).astype(int)
df["semirechazo"] = (df["j0900b"] == 1).astype(int)
df["nopiden"] = (df["j0900c"] == 1).astype(int)

df["alguna"] = ((df["rechazo"]==1) | 
                (df["semirechazo"]==1) | 
                (df["nopiden"]==1)).astype(int)

# -----------------------------
# Check
print(df[["ahorra","ahorros","cred_inf","cred_line","cred_add",
          "impago","ayudas","rechazo","semirechazo","nopiden","alguna"]].head())



df.replace([-97, -98, -99, -5], pd.NA, inplace=True)
```
