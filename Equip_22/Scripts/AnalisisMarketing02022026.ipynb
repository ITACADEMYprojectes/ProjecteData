```python
import os
import pandas as pd
```

```python
# cargar datos
ruta = os.path.join("..", 'Data', 'TouristAccommodationClean02022026.csv')
df = pd.read_csv(ruta)
```

```python
df = df[df['neighbourhood_name'] != 'No Asignado']
```

```python
neighbourhood_stats = df.groupby(['city', 'neighbourhood_name']).agg(
    mean_review_rating=('review_scores_rating', 'mean'),
    mean_review_location=('review_scores_location', 'mean'),
    median_max_nights=('maximum_nights', 'median'),
    median_min_nights=('minimum_nights', 'median')
    ).reset_index()
```

```python
valoraciones = {'mean_review_rating': 'Reseñas Total Medias',
                'mean_review_location': 'Puntuaciones Zona Medias',
                'median_max_nights': 'Noches Máximas Medianas',
                'median_min_nights': 'Noches Mínimas Medianas'
                }
```

```python
for categoria, nombre in valoraciones.items():
    if categoria != 'median_min_nights':
        ranking = neighbourhood_stats.sort_values(by=categoria).head(5)
        ranking = ranking[['city', 'neighbourhood_name', categoria]]
        print(f'Los 5 barrios con {nombre} más bajas son:\n{ranking.to_string(index=False)}\n')
    else:
        ranking = neighbourhood_stats.sort_values(by=categoria, ascending=False).head(5)
        ranking = ranking[['city', 'neighbourhood_name', categoria]]
        print(f'Los 5 barrios con {nombre} más altas son:\n{ranking.to_string(index=False)}\n')
```
