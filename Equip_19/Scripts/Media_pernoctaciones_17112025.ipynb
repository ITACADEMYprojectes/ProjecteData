```python
import pandas as pd
import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("datos_INE_consolidados_17112025.csv")
df
```

```python
df["media_pernoctaciones"].describe()
```

```python
df["provincia"].unique()
```

```python
evo_pern_tot= df.groupby(["fecha"])[["media_pernoctaciones"]].sum()
evo_pern_tot
```

```python
fig = px.line(evo_pern_tot, y = ["media_pernoctaciones"], title = "Evolución de la media de pernoctaciones", 
              color_discrete_map={"Media_pernoctaciones": "#ff0000"}, line_dash= "variable", line_dash_map = {'media_pernoctaciones': 'solid'})
fig.update_layout(title_y=0.01, title_x = 0.5)
fig.show()
```

```python
df = df.sort_values(by="fecha")
df_fecha = df.set_index("fecha")
df_fecha.reset_index(inplace=True)
df_fecha
```

```python
df_fecha["fecha"] = pd.to_datetime(df_fecha["fecha"])
```

```python
df_fecha.info()
```

```python
avg_pern_prov = df_fecha.groupby(["provincia", pd.Grouper(key="fecha", freq="MS")])[["media_pernoctaciones", "pais_residencia"]].mean()
avg_pern_prov
```

```python
avg_pern_prov = avg_pern_prov.reset_index(level=0)
avg_pern_prov
```

```python
fig = px.line(avg_pern_prov, y = "media_pernoctaciones", title = "Evolución de la media de pernoctaciones", color="provincia")
fig.update_layout(title_y=0.01, title_x = 0.5)
fig.show()
```

```python
fig = px.line(avg_pern_prov[avg_pern_prov['provincia']=="Sevilla"], y = "media_pernoctaciones", title = "Evolución de la media de pernoctaciones", color="provincia")
fig.update_layout(title_y=0.01, title_x = 0.5)
fig.show()
```
