# Quin percentatge d'allotjaments tenen una valoració general més gran de 80 a cada ciutat?

```python
import pandas as pd

# Creamos una columna nueva que será TRUE si el review_scores_rating es superior a 80
df["rating_greather_than_80"] = df["review_scores_rating"] > 80

df.head()
```

```python
# Agrupamos por ciudad y calculamos el porcentaje
df_grouped = (df.groupby("city")["rating_greather_than_80"].mean() * 100).round(2).reset_index()

df_grouped
```

```python
# Ordenamos por porcentaje en orden desc
df_sorted = df_grouped.sort_values(by="rating_greather_than_80", ascending=False).reset_index(drop=True)
```

```python
# Renombramos la columna
df_sorted = df_sorted.rename(columns={"rating_greather_than_80": "%_rating_greather_than_80"})

df_sorted
```

```python
def calculate_apartment_ratings_by_city(df):
    """
    Calcula estadísticas de apartamentos por ciudad basadas en puntuaciones de reseñas.
    
    Args:
        df (pandas.DataFrame): DataFrame con columnas 'city', 'apartment_id', 'review_scores_rating'
    
    Returns:
        pandas.DataFrame: DataFrame con las estadísticas por ciudad
    """
    # Filtrar registros con review_scores_rating no nulo
    filtered_df = df[df['review_scores_rating'].notnull()].copy()
    
    # Calcular si el rating es mayor a 80 (después de dividir por 10)
    filtered_df['above_80'] = (filtered_df['review_scores_rating']  > 80).astype(int)
    
    # Agrupar por ciudad y calcular las métricas
    result = (
        filtered_df
        .groupby('city')
        .agg(
            total_apartments=('apartment_id', 'count'),
            apartments_above_80=('above_80', 'sum')
        )
        .assign(
            percentage_above_80=lambda x: (x['apartments_above_80'] / x['total_apartments']) * 100
        )
        .reset_index()
    )
    
    return result


result = calculate_apartment_ratings_by_city(df)
print(result)
```
