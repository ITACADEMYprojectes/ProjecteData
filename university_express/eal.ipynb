```python
import pandas as pd

eal_2023_dict = pd.read_excel("Tablas_EAL_2023.xlsx", sheet_name=["EAL-18a"], skiprows=5)

df_18a = eal_2023_dict["EAL-18a"]

df_18a.columns = [
    "Categoria", 
    "Total", 
    "Cursos_Externos", 
    "Cursos_Internos", 
    "Formacion_Puesto", 
    "Rotacion_Puestos", 
    "Conferencias_Seminarios"
]


df_18a = df_18a.dropna(subset=["Categoria"])

df_18a = df_18a[df_18a["Total"].notna()]

df_18a
```

```python
# 1. Ponemos la columna 'Categoria' como el índice (los nombres de las filas)
df_transpuesta = df_18a.set_index("Categoria")

# 2. Transponemos (la .T hace la magia de girar todo)
df_transpuesta = df_transpuesta.T

# Ahora tus columnas son "TOTAL", "De 5 a 9 trabajadores", "Andalucía", etc.
# Y tus filas son "Total", "Cursos_Externos", "Cursos_Internos"...
```

```python
df_transpuesta
```

```python
# 1. Quitamos los espacios en blanco al principio y al final de todos los nombres de columnas
df_transpuesta.columns = df_transpuesta.columns.str.strip()

# 2. Ahora ya puedes usar los nombres "limpios" sin miedo al KeyError
columnas_interes = [
    "De 250 a 499 trabajadores", 
    "Cataluña", 
    "Andalucía", 
    "Madrid (Comunidad de)", 
    "País Vasco", 
    "Comunitat Valenciana"
]

# 3. Filtramos
df_seleccion = df_transpuesta[columnas_interes]

print(df_seleccion)
```

```python
# Tienes que poner el "=" para que se guarde el cambio
df_seleccion = df_seleccion.round(2)

# Ahora, al imprimirla, verás que todo tiene 2 decimales
print(df_seleccion)
```

```python
# 1. Convertimos todo a números (por si acaso Pandas cree que es texto)
df_seleccion = df_seleccion.apply(pd.to_numeric, errors='coerce')

# 2. Ahora redondeamos y GUARDAMOS el resultado
df_seleccion = df_seleccion.round(2)

# 3. Comprobamos
print(df_seleccion)
```

```python
# Ahora 'df_formacion' contiene exactamente lo mismo que 'eal_2023_clean'
# 1. Creamos el nuevo nombre usando la tabla que SÍ existe
eal_2023_18a = df_seleccion

# 2. Comprobamos que funciona

```
