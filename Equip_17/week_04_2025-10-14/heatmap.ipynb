```python
import pandas as pd

df_test = pd.read_csv("data/viajeros_pernoctaciones_estancia.csv")

# Paso 2: Crear un nuevo DataFrame con las columnas deseadas
df_subset = df_test[['periodo', 'provincia', 'estancia_media_total']]

# Paso 3: Mostrar las primeras filas
print(df_subset.head())
```

```python
# Paso 4: Filtrar solo los registros del año 2024
df_2024 = df_subset[df_subset['periodo'].astype(str).str.startswith('2024')]

# Mostrar las primeras filas filtradas
df_2024
```

```python
# Paso 5 (versión limpia): crear una columna 'mes' solo con el nombre del mes
meses = {
    'M01': 'enero',
    'M02': 'febrero',
    'M03': 'marzo',
    'M04': 'abril',
    'M05': 'mayo',
    'M06': 'junio',
    'M07': 'julio',
    'M08': 'agosto',
    'M09': 'septiembre',
    'M10': 'octubre',
    'M11': 'noviembre',
    'M12': 'diciembre'
}

# Extraer el código de mes y mapearlo a nombre
df_2024 = df_2024.copy()  # evita el warning
df_2024['mes'] = df_2024['periodo'].str.extract(r'(M\d{2})')[0].map(meses)

# Mostrar resultado
df_2024
```

```python
# Paso 5 (versión limpia): crear una columna 'mes' solo con el nombre del mes
meses = {
    'M01': 'enero',
    'M02': 'febrero',
    'M03': 'marzo',
    'M04': 'abril',
    'M05': 'mayo',
    'M06': 'junio',
    'M07': 'julio',
    'M08': 'agosto',
    'M09': 'septiembre',
    'M10': 'octubre',
    'M11': 'noviembre',
    'M12': 'diciembre'
}

# Extraer el código de mes y mapearlo a nombre
df_2024 = df_2024.copy()  # evita el warning
df_2024['mes'] = df_2024['periodo'].str.extract(r'(M\d{2})')[0].map(meses)

# Mostrar resultado
df_2024[['mes', 'provincia', 'estancia_media_total']].head()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1) Orden cronológico de meses
orden_meses = [
    'enero','febrero','marzo','abril','mayo','junio',
    'julio','agosto','septiembre','octubre','noviembre','diciembre'
]
df_2024 = df_2024.copy()
df_2024['mes'] = pd.Categorical(df_2024['mes'], categories=orden_meses, ordered=True)

# 2) Tabla provincias x meses con la media de estancia
tabla = df_2024.pivot_table(
    index='provincia',
    columns='mes',
    values='estancia_media_total',
    aggfunc='mean'
)

# 3) Ordenar provincias por media anual
tabla = tabla.reindex(tabla.mean(axis=1).sort_values(ascending=False).index)
tabla = tabla.reindex(columns=orden_meses)

# 4) Crear el heatmap
fig, ax = plt.subplots(figsize=(12, 10))
im = ax.imshow(tabla.values, aspect='auto', cmap='viridis')

# Ejes y etiquetas
ax.set_xticks(range(len(tabla.columns)))
ax.set_xticklabels(tabla.columns, rotation=45, ha='right')
ax.set_yticks(range(len(tabla.index)))
ax.set_yticklabels(tabla.index)

ax.set_xlabel('Mes')
ax.set_ylabel('Provincia')
ax.set_title('Estancia media 2024 (noches)')

# Barra de color
cbar = plt.colorbar(im, ax=ax)
cbar.set_label('Estancia media (noches)')

# 5) Añadir los valores encima de los colores
for i in range(len(tabla.index)):
    for j in range(len(tabla.columns)):
        valor = tabla.iloc[i, j]
        if not np.isnan(valor):
            ax.text(j, i, f"{valor:.2f}", ha='center', va='center', color='white', fontsize=8)

plt.tight_layout()
plt.show()
```

```python
# Convertir tabla pivote a formato largo para Flourish
tabla_larga = tabla.reset_index().melt(
    id_vars='provincia',
    var_name='mes',
    value_name='estancia_media_total'
)

# Guardar lista larga
tabla_larga.to_csv("data/estancia_media_2024_flourish.csv", index=False, encoding='utf-8-sig')

print("✅ Archivo guardado: estancia_media_2024_flourish.csv")
```
