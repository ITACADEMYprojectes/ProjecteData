```python
# Carga del dataset WK1 para comparativa evolutiva
WK1_PATH = "../Data/2026-06-29_1_bank_dataset_raw.csv"
df_wk1 = pd.read_csv(WK1_PATH)

# Identificación de registros nuevos en WK2
ids_wk1 = set(df_wk1["id"])
ids_wk2 = set(df["id"])
ids_nuevos = ids_wk2 - ids_wk1
df_nuevos = df[df["id"].isin(ids_nuevos)].copy()

print(f"Registros WK1 : {len(df_wk1):,}")
print(f"Registros WK2 : {len(df):,}")
print(f"Registros nuevos WK2 : {len(df_nuevos):,}")
print(f"\nDistribución deposit en registros nuevos:")
print(df_nuevos["deposit"].value_counts())
```

```python
# Funciones reutilizables — relación variable vs deposit


def analyze_numeric_target(data, variable, target="deposit"):
    """Estadísticos descriptivos y boxplot de una variable numérica por deposit."""
    summary = (
        data.groupby(target)[variable]
        .agg(Recuento="count", Media="mean", Mediana="median",
             Desv_Std="std", Mínimo="min", Máximo="max")
        .round(2)
    )
    print(f"Variable numérica: {variable}")
    display(summary)

    plt.figure(figsize=(10, 6))
    sns.boxplot(data=data, x=target, y=variable, hue=target,
                palette=["#1F5C45", "#9FC2A9"], legend=False)
    plt.title(f"{variable.capitalize()} según contratación de depósito")
    plt.xlabel("deposit")
    plt.ylabel(variable.capitalize())
    plt.tight_layout()
    plt.show()
    return summary


def analyze_categorical_target(data, variable, target="deposit"):
    """Tabla de porcentajes y gráfico de barras apiladas por deposit."""
    tabla = (
        pd.crosstab(data[variable], data[target], normalize="index") * 100
    ).round(2)
    print(f"Variable categórica: {variable} — % por fila")
    display(tabla)

    ax = tabla.plot(kind="bar", stacked=True, figsize=(11, 6),
                    color=["#1F5C45", "#9FC2A9"])
    plt.title(f"{variable.capitalize()} vs deposit")
    plt.xlabel(variable.capitalize())
    plt.ylabel("Porcentaje (%)")
    plt.legend(title="deposit", bbox_to_anchor=(1.05, 1), loc="upper left")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
    return tabla


def tasa_deposito_si(data, variable):
    """Calcula el % de deposit='yes' por categoría de una variable."""
    return (
        data.groupby(variable)["deposit"]
        .apply(lambda x: (x == "yes").mean() * 100)
        .round(2)
        .rename("tasa_deposito_si_%")
    )


def comparar_tasas_wk1_wk2(variable, orden=None):
    """Tabla comparativa de tasas de deposit='yes' entre WK1 y WK2."""
    t1 = tasa_deposito_si(df_wk1, variable)
    t2 = tasa_deposito_si(df, variable)
    comp = pd.DataFrame({"WK1_%": t1, "WK2_%": t2})
    comp["Δ_pp"] = (comp["WK2_%"] - comp["WK1_%"]).round(2)
    if orden:
        comp = comp.reindex(orden)
    comp = comp.sort_values("WK2_%", ascending=False)
    display(comp.round(2))
    return comp


def resumen_objetivo(data, etiqueta):
    """Distribución de deposit en un dataset."""
    res = data["deposit"].value_counts(normalize=True).mul(100).round(2)
    res.index.name = "deposit"
    print(f"Distribución deposit — {etiqueta}")
    display(res.rename("%"))
    return res
```
