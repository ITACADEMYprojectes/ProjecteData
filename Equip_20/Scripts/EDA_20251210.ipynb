### Visualizaciones EDA

```python
# Importamos las librerías a emplear en las visualizaciones
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
# Configuración básica para los gráficos
sns.set_style("whitegrid")
plt.rcParams['figure.dpi'] = 100 # Mejora de la resolución de los gráficos
# Paleta de colores Seaborn Set2, para usarla con MatPlotLib y seguir el mismo estilo
sns_color_set2 = ["#66c2a5", "#fc8d62", "#8da0cb", "#e78ac3", "#a6d854", "#ffd92f", "#e5c494", "#b3b3b3"]
```

##### Gráfico de Empresas Formadoras por Tamaño, Actividad y Comunidad

```python
# Crear el gráfico de lineas
plt.figure(figsize=(12, 6))
sns.lineplot(
    data=df_empresas_formadoras.melt(id_vars='Año', var_name='Categoría', value_name='Porcentaje'),
    x='Año',
    y='Porcentaje',
    hue='Categoría',
    palette='Set2'
)

# Añadir etiquetas y título
plt.title('Empresas Formadoras por Tamaño, Actividad y Comunidad', fontsize=16)
plt.xlabel('Año', fontsize=14)
plt.ylabel('Porcentaje sobre el total de empresas', fontsize=14)

# Rotar las etiquetas del eje X para que sean legibles
plt.xticks(rotation=45, ha='right')

plt.tight_layout() # Ajustar el diseño para que todo quepa
plt.show()
```

##### Gráfico de Necesidades Formativas por Tamaño, Actividad y Comunidad

```python
# Crear el gráfico de barras
plt.figure(figsize=(12, 6))
ax = sns.barplot(
    data=df_necesidades.melt(id_vars='Año', var_name='Categoría', value_name='Porcentaje'),
    x='Año',
    y='Porcentaje',
    hue='Categoría',
    palette='Set2'
)

# Añadir las etiquetas directamente sobre las barras (filtrar patches que no tienen get_height)
for p in ax.patches:
    # Algunos elementos en ax.patches no tienen get_height (p.ej. PathPatch u otros), por eso verificamos
    if not hasattr(p, "get_height"):
        continue
    height = p.get_height() # type: ignore
    # Evitar anotar valores NaN
    if np.isnan(height):
        continue
    ax.annotate(f'{height:.2f}%',
                xy=(p.get_x() + p.get_width() / 2, height), # type: ignore
                xytext=(0, 3),
                textcoords='offset points',
                ha='center', va='bottom',
                fontsize=9, color='dimgray')

# Añadir etiquetas y título
plt.title('Necesidades Formativas por Tamaño de la empresa, Actividad y Comunidad', fontsize=16)
plt.xlabel('Año', fontsize=14)
plt.ylabel('Porcentaje sobre el total de empresas', fontsize=14)

# Rotar las etiquetas del eje X para que sean legibles
plt.xticks(rotation=45, ha='right')

plt.tight_layout() # Ajustar el diseño para que todo quepa
plt.show()
```

##### Gráfico de serie temporal de Competencias Formativas

```python
#  Reestructurar los datos a formato "largo" (tidy data)
# Este paso es CRUCIAL para usar Seaborn FacetGrid o catplot/relplot de manera efectiva.
df_long = df_competencias.melt(
    id_vars=['Año', 'Tipo'],
    var_name='Categoría',
    value_name='Porcentaje'
)

# Definición de la visualización robusta (Small Multiples con Seaborn)

# Configuración del estilo
sns.set_theme(style="whitegrid")

# Creación de los múltiples gráficos de línea (Small Multiples)
# relplot es ideal para series temporales y usa FacetGrid para crear el layout.
g = sns.relplot(
    data=df_long,
    x='Año',
    y='Porcentaje',
    # 'col' divide los gráficos por 'Tipo' de competencia
    col='Tipo',
    # 'hue' traza las líneas individuales para cada 'Categoría'
    hue='Categoría',
    kind='line',
    # Definición de la cuadrícula (e.g., 2 filas para los 5 tipos)
    col_wrap=3,
    height=4, 
    aspect=1.2, # Ajuste de la proporción ancho/alto
    palette='Set2',
    marker='o', # Añadir marcadores para visualizar los puntos de datos discretos (años)
    dashes=False # Usar líneas sólidas
)

# Personalización del gráfico

# Título general
g.figure.suptitle('Evolución Anual (2018-2023) de los Tipos de Competencia por Categoría', 
                  fontsize=16, fontweight='bold', y=1.03)

# Etiquetas y formato de los ejes
for ax in g.axes.flat:
    # Formato de eje X para mostrar años completos
    ax.set_xticks(df_competencias['Año'].unique())
    ax.set_xticklabels(df_competencias['Año'].unique().astype(str), rotation=45, ha='right')
    
    # Añadir el símbolo de porcentaje al eje Y
    ax.set_ylabel('Porcentaje (%)', fontsize=10)
    ax.set_xlabel('Año', fontsize=10)
    
    # Ajustar el título de cada subgráfico (el 'Tipo' de competencia)
    new_title = ax.get_title().split(' = ')[-1]
    ax.set_title(new_title, fontsize=12, fontweight='semibold')
    
    # Mejorar la leyenda (se añade automáticamente por relplot)
    if ax.get_legend() is not None:
        ax.get_legend().set_title('Categoría')
        
# Ajustar el layout
plt.tight_layout(rect=(0, 0, 1, 0.98)) # Usar tuple para evitar error de tipos y dejar espacio al suptitle

# Mostrar el gráfico
plt.show()
```

##### Gráfico de barras agrupadas de Medios Formativos por Tipo y Categorías

```python
# Reestructurar los datos a formato "largo" (tidy data) para facilitar la visualización
# 'melt' transforma las columnas de categorías en filas.
df= df_medios.melt(
    id_vars='Tipo',
    var_name='Categorías',
    value_name='Porcentaje'
)

# 3. Definición de la visualización robusta con Matplotlib

# Configuración del estilo
plt.style.use('seaborn-v0_8-whitegrid') # Un estilo limpio y profesional

# Creación de la figura y los ejes
fig, ax = plt.subplots(figsize=(12, 7))

# Definición de la posición de las barras
tipos = df['Tipo'].unique()
x = np.arange(len(tipos)) # Posiciones de las etiquetas 'Tipo'
width = 0.25 # Ancho de las barras

# Nombres de las categorías (columnas a graficar)
categorias = ['Porcentaje_x_tamaño', 'Porcentaje_x_A_E', 'Catalunya']

# Generación de las barras para cada categoría
# Utilizamos un bucle for para automatizar el desplazamiento de las barras (offset)
for i, cat in enumerate(categorias):
    # Extraer los datos de la categoría actual
    porcentajes = df[df['Categorías'] == cat]['Porcentaje']
    
    # Calcular la posición de las barras (desplazamiento para agrupar)
    offset = (i - 1) * width
    
    # Dibujar las barras
    rects = ax.bar(x + offset, porcentajes, width, label=cat, zorder=3)
    
    # Añadir las etiquetas de valor (data labels) para mayor precisión
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.1f}%',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),  # 3 puntos de desplazamiento vertical
                    textcoords="offset points",
                    ha='center', va='bottom',
                    fontsize=9,
                    color='dimgray')

# 4. Personalización del gráfico

# Título y etiquetas
ax.set_title('Distribución de Medios Formativos por Tipo y Categorías (2023)', 
             fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Tipo de Medio', fontsize=12)
ax.set_ylabel('Porcentaje sobre el total de empresas', fontsize=12)

# Etiquetas del eje X
ax.set_xticks(x)
ax.set_xticklabels(tipos, rotation=45, ha="right", fontsize=10) # Rotación para mejor legibilidad

# Leyenda
ax.legend(title='Categoría', loc='upper right', bbox_to_anchor=(1.0, 1.0))

# Ajustar diseño para evitar el recorte de etiquetas
plt.tight_layout()

# Mostrar el gráfico
plt.show()
```
