# Carga de los Datos

## imports

```python
import sqlalchemy
import pandas as pd
```

## SQL conexión

```python
engine = sqlalchemy.create_engine("mysql+pymysql:///Equip_30", 
                                  connect_args={"read_default_file": "db_config.cnf"})
```

```python
try:
    with engine.connect() as conn:
        print("Conexión exitosa")
        inspector = sqlalchemy.inspect(engine)
        tablas = inspector.get_table_names()
        
        for tabla in tablas:
            df = pd.read_sql(f"SELECT * FROM {tabla};", conn)
            df.to_csv(f"../Data/{tabla}_raw.csv", index=False) # COPIA RAW

except:
    print("Error al conectar con MySQL")  
```
